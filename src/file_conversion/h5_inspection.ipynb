## Code to inspect h5 files, change filepath to a h5 file you have

```python
import h5py

def inspect_h5_file(filepath):
    with h5py.File(filepath, 'r') as f:
        print(f"Inspecting: {filepath}")
        print("\n📁 Top-level groups/datasets:")
        f.visititems(lambda name, obj: print(f" - {name}: {type(obj).__name__}"))

        print("\n📐 Dataset info:")
        for name, obj in f.items():
            if isinstance(obj, h5py.Dataset):
                print(f" - {name}: shape={obj.shape}, dtype={obj.dtype}")
            elif isinstance(obj, h5py.Group):
                for k in obj.attrs:
                    print(f" - metadata/{k} = {obj.attrs[k]}")

        print("\n✅ Done.")

# Change the filepath to your HDF5 file below
filepath = r"C:\Users\thoma\Code\ResearchCode\spike_to_h5\ekg_resp_data\sub_5_13_baseline_20250523_105046.rec\sub_5_13_baseline_20250523_105046_merged.h5"
d13 = r'C:\Users\thoma\Code\ResearchCode\respiratory_pilot\src\file_conversion\ekg_resp_data\subj_5_13_baseline_d13_20250526_143515.rec\subj_5_13_baseline_d13_20250526_143515_merged.h5'
manual_conv_path = r'C:\Users\thoma\Code\ResearchCode\respiratory_pilot\src\file_conversion\h5_outputs\sub_5_13_juvenile_20250523_110554_merged.h5'

print("Not working")
inspect_h5_file(filepath)

print("\nUsed to work")
inspect_h5_file(d13)

print("\nManual conversion")
inspect_h5_file(manual_conv_path)
```
