```python
import os
import h5py

def inspect_h5_file(filepath):
    print(f"\n🔍 Inspecting: {os.path.basename(filepath)}")
    try:
        with h5py.File(filepath, 'r') as f:
            print("📁 Top-level groups/datasets:")
            f.visititems(lambda name, obj: print(f" - {name}: {type(obj).__name__}"))

            print("\n📐 Dataset info:")
            for name, obj in f.items():
                if isinstance(obj, h5py.Dataset):
                    print(f" - {name}: shape={obj.shape}, dtype={obj.dtype}")
                elif isinstance(obj, h5py.Group):
                    for k in obj.attrs:
                        print(f" - metadata/{k} = {obj.attrs[k]}")
    except Exception as e:
        print(f"❌ Error reading file: {e}")
    print("✅ Done.\n" + "-"*50)

# 🔁 Loop through all .h5 files in the specified folder
folder = r"C:\Users\thoma\Code\ResearchCode\respiratory_pilot\src\file_conversion\day1_h5_files"

if not os.path.exists(folder):
    print(f"❌ Folder does not exist: {folder}")
else:
    h5_files = [f for f in os.listdir(folder) if f.endswith('.h5')]
    print(f"\n📦 Found {len(h5_files)} HDF5 files in {folder}\n")

    for fname in h5_files:
        filepath = os.path.join(folder, fname)
        inspect_h5_file(filepath)
```

```python
import os
import h5py
import matplotlib.pyplot as plt
import numpy as np

def plot_ekg_resp_from_h5(filepath, downsample_factor=100):
    try:
        with h5py.File(filepath, 'r') as f:
            print(f"\n📈 Plotting: {os.path.basename(filepath)}")
            ekg = f['ekg'][:]
            resp = f['resp'][:]
            meta = dict(f['metadata'].attrs)

        # Downsample for faster plotting
        ekg_ds = ekg[::downsample_factor]
        resp_ds = resp[::downsample_factor]

        plt.figure(figsize=(12, 4))
        plt.suptitle(os.path.basename(filepath), fontsize=10)

        plt.subplot(2, 1, 1)
        plt.plot(ekg_ds)
        plt.title("EKG (Downsampled)", fontsize=9)

        plt.subplot(2, 1, 2)
        plt.plot(resp_ds)
        plt.title("Respiration (Downsampled)", fontsize=9)

        plt.tight_layout(rect=[0, 0, 1, 0.95])
        plt.show()
    except Exception as e:
        print(f"❌ Failed to plot {filepath}: {e}")

# 🔁 Loop through all .h5 files in the specified folder
folder = r"C:\Users\thoma\Code\ResearchCode\respiratory_pilot\src\file_conversion\day1_h5_files"

if not os.path.exists(folder):
    print(f"❌ Folder does not exist: {folder}")
else:
    h5_files = [f for f in os.listdir(folder) if f.endswith('.h5')]
    print(f"\n📂 Found {len(h5_files)} HDF5 files\n")

    for fname in h5_files:
        filepath = os.path.join(folder, fname)
        plot_ekg_resp_from_h5(filepath)
```

```python
import h5py
import numpy as np
import matplotlib.pyplot as plt

# ✅ Path to one of your .h5 files
filepath = r"C:\Users\thoma\Code\ResearchCode\respiratory_pilot\src\file_conversion\day1_h5_files\BLRI_s2_3_p5_3_nRB3_20250622_101813_merged.h5"

with h5py.File(filepath, 'r') as f:
    resp = f['resp'][:]
    meta = dict(f['metadata'].attrs)

    print(resp.dtype)


# 🧼 Flatten if needed (e.g., if shape is (N, 1) or (1, N))
if resp.ndim > 1:
    print(f"Original shape: {resp.shape} — flattening.")
    resp = resp.flatten()
else:
    print(f"Loaded RESP shape: {resp.shape}")

# Optional smoothing
window = 500
if len(resp) > window:
    kernel = np.ones(window) / window
    smoothed_resp = np.convolve(resp, kernel, mode='valid')
else:
    print("⚠️ Not enough samples to smooth — using raw.")
    smoothed_resp = resp

# 🫁 Plot full resp trace
plt.figure(figsize=(16, 4))
plt.plot(smoothed_resp, linewidth=0.8)
plt.title(f"Full Resp Trace — Subject {meta.get('subject_id', '?')} on {meta.get('date', '?')}")
plt.xlabel("Sample Index")
plt.ylabel("Voltage")
plt.grid(True)
plt.tight_layout()
plt.show()
```
