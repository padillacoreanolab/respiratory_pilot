```python
import cv2

cap = cv2.VideoCapture(input_video)
print(cap.isOpened())
```

```python
output_video = r"D:\BLA_ChR_resp\BLA_ChR_resp_RI1_hc_day2\4_2_RI1_f2_g_on_20260111_130824.rec\4_2_RI1_f2_g_on_20260111_130824.1_16fps.mp4"
```

```python
original_fps = cap.get(cv2.CAP_PROP_FPS)
frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

print("Original FPS tag:", original_fps)
print("Total frames:", frame_count)
```

```python
true_fps = 16.0

width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

fourcc = cv2.VideoWriter_fourcc(*"mp4v")
out = cv2.VideoWriter(output_video, fourcc, true_fps, (width, height))
```

```python
while True:
    ret, frame = cap.read()
    if not ret:
        break
    out.write(frame)
```

```python
cap.release()
out.release()

print("✅ Done! 16 fps video created.")
```
