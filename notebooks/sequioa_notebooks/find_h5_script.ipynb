```python
import os
import glob

def find_h5_files(root_dir, prefix):
    """
    Recursively finds all .h5 files in a directory and returns a dict
    where keys are the file names (without extension) and values are full paths.
    """
    h5_paths = {}
    for file in glob.glob(os.path.join(root_dir, "**", "*.h5"), recursive=True):
        key = os.path.splitext(os.path.basename(file))[0]
        if prefix.lower() in key.lower():  # filter if you want only CM or BL
            h5_paths[key] = file
    print(f"✅ Found {len(h5_paths)} {prefix.upper()} .h5 files")
    return h5_paths


# Example usage
resp_paths_bl = find_h5_files(r"E:\Aim1\AIM1\Cagemate_bl_no_ISO\h5_outputs", "BL")
resp_paths_cm = find_h5_files(r"E:\Aim1\AIM1\Day1_new\resp_CM_h5", "CM")
```

```python
def find_boris_files(root_dir, prefix="CM"):
    """
    Finds all .csv BORIS files in a directory.
    """
    csv_paths = {}
    for file in glob.glob(os.path.join(root_dir, "**", "*.csv"), recursive=True):
        key = os.path.splitext(os.path.basename(file))[0]
        if prefix.lower() in key.lower():
            csv_paths[key] = file
    print(f"✅ Found {len(csv_paths)} {prefix.upper()} BORIS files")
    return csv_paths


boris_paths_cm = find_boris_files(r"E:\Aim1\AIM1\Day1_new\cm_boris", "CM")
boris_paths_bl = find_boris_files(r"E:\Aim1\AIM1\Day1_new\cm_boris", "BL")

```
