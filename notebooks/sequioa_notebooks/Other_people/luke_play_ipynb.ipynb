```python
def load_clean_resp_signal(h5_file, target_rate=100):
    """
    Load respiration from .h5, anti-alias LPF at raw fs, resample to target_rate,
    band-pass clean, and return (signal, time, fs_out).
    """
    try:
        with h5py.File(h5_file, 'r') as f:
            resp = f['resp'][:].flatten()
            ekg_meta = dict(f['ekg_metadata'].attrs)
        duration_sec = float(ekg_meta['duration_sec'])
        fs = len(resp) / duration_sec
    except Exception as e:
        print(f"Error loading {h5_file}: {e}")
        return None, None, None

    # --- Anti-alias LPF at raw fs ---
    if target_rate >= fs:
        # No decimation needed; still low-pass a bit to remove HF noise
        cutoff_hz = min(0.45 * fs, 20.0)  # keep it sensible if fs is small
        Wn = cutoff_hz / (fs / 2.0)
    else:
        cutoff_hz = target_rate / 2.0       # classic anti-alias cutoff
        Wn = cutoff_hz / (fs / 2.0)

    Wn = min(max(Wn, 0.0001), 0.9999)       # clamp for safety
    b, a = butter(N=4, Wn=Wn, btype='low')
    filtered = filtfilt(b, a, resp)

    # --- Resample to EXACT target_rate ---
    # Prefer exact-length resample to avoid floor issues
    N_out = int(round(duration_sec * target_rate))
    # If you prefer polyphase, use Fraction to get integers:
    # from fractions import Fraction
    # frac = Fraction(target_rate, int(round(fs))).limit_denominator()
    # resampled = resample_poly(filtered, up=frac.numerator, down=frac.denominator)
    resampled = scipy.signal.resample(filtered, N_out)

    # --- Final band-pass clean at the new fs ---
    # Keep physiological band; adjust if you know your spectrum well
    lowcut, highcut = 0.1, 15.0
    rsp_cleaned = nk.signal_filter(
        resampled, lowcut=lowcut, highcut=highcut,
        method="butterworth", sampling_rate=target_rate, order=2
    )

    time_vector = np.arange(len(rsp_cleaned)) / float(target_rate)  # ✅ correct
    return rsp_cleaned, time_vector, float(target_rate)
```

```python
from scipy.signal import find_peaks

def compute_bout_resp_rate(signal, time, t_start, t_stop):
    """
    Compute respiration rate (Hz) within a bout window.
    """
    mask = (time >= t_start) & (time <= t_stop)
    if mask.sum() < 10:
        return np.nan

    seg_signal = signal[mask]
    seg_time = time[mask]

    peaks, _ = find_peaks(seg_signal, distance=20)  # distance tuned to fs
    duration = seg_time[-1] - seg_time[0]

    if duration <= 0:
        return np.nan

    return len(peaks) / duration
```

```python
processed_boris = {}

for sess, paths in sessions.items():
    print(f"\nProcessing {sess}")

    if paths["boris"] is None:
        print("  ⚠️ No BORIS file, skipping")
        continue

    # ---- Load + clean respiration ----
    signal, time, fs = load_clean_resp_signal(paths["h5"])
    if signal is None:
        print("  ❌ Resp failed to load")
        continue

    boris_df = paths["boris"].copy()

    # ---- Sanity check BORIS columns ----
    print("  BORIS columns:", boris_df.columns.tolist())

    # EXPECTED: Start (s), Stop (s)
    assert "Start (s)" in boris_df.columns
    assert "Stop (s)" in boris_df.columns

    # ---- Compute bout respiration rate ----
    bout_rates = []
    for _, row in boris_df.iterrows():
        rate = compute_bout_resp_rate(
            signal,
            time,
            row["Start (s)"],
            row["Stop (s)"]
        )
        bout_rates.append(rate)

    boris_df["BoutRespRate"] = bout_rates
    boris_df["Session"] = sess

    processed_boris[sess] = boris_df

    print(f"  ✅ Added BoutRespRate to {sess}")
```

```python
import os
import h5py

resp_out_dir = r"E:\Aim1\AIM1\Day1_new\SHARE_clean_resp_and_boris\resp_clean_h5"
os.makedirs(resp_out_dir, exist_ok=True)

for sess, paths in sessions.items():
    signal, time, fs = load_clean_resp_signal(paths["h5"])
    if signal is None:
        continue

    out_path = os.path.join(resp_out_dir, f"{sess}_clean_resp.h5")

    with h5py.File(out_path, "w") as f:
        f.create_dataset("resp_signal", data=signal)
        f.create_dataset("time", data=time)
        f.attrs["sampling_rate"] = fs
        f.attrs["preprocessed"] = True
        f.attrs["notes"] = "Cleaned respiration only (no alignment applied)"

    print(f"Saved clean respiration → {out_path}")
```
