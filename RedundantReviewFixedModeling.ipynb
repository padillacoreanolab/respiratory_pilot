```python
import numpy as np
from scipy.signal import find_peaks

def align_respiratory_behavioral_data(bouts, time_vec, signal_data, rates_per_sec, rates_per_min, intervals, interval_times, sampling_rate=100):
    results = []
    
    for idx, bout in bouts.iterrows():
        start = bout['start_time']
        end = bout['end_time']
        
        # Extract respiratory signal for this bout
        signal_mask = (time_vec >= start) & (time_vec <= end)
        bout_signal = signal_data[signal_mask]
        bout_time = time_vec[signal_mask]
        
        # Default values
        sig_mean = sig_min = sig_max = sig_range = np.nan
        mean_rate_sec = mean_rate_min = mean_interval = np.nan
        breath_count = 0
        ibi = []
        inst_rates = []
        
        if len(bout_signal) > 0:
            # Basic respiratory signal stats
            sig_mean = np.mean(bout_signal)
            sig_min = np.min(bout_signal)
            sig_max = np.max(bout_signal)
            sig_range = sig_max - sig_min

            # --- NEW: Instantaneous breathing rate calculation ---
            peaks, _ = find_peaks(bout_signal, distance=sampling_rate * 0.0833)  # at least 0.0833s apart (12 Hz max)
            peak_times = peaks / sampling_rate  # convert to seconds
            if len(peak_times) > 1:
                ibi = np.diff(peak_times)                 # inter-breath intervals
                inst_rates = 1.0 / ibi                    # instantaneous rate (Hz)
        
        # Breathing rate from peak intervals (your existing code)
        rate_mask = (interval_times >= start) & (interval_times <= end)
        bout_rates_sec = rates_per_sec[rate_mask]
        bout_rates_min = rates_per_min[rate_mask]
        bout_intervals = intervals[rate_mask]
        
        if len(bout_rates_sec) > 0:
            mean_rate_sec = np.mean(bout_rates_sec)
            mean_rate_min = np.mean(bout_rates_min)
            mean_interval = np.mean(bout_intervals)
            breath_count = len(bout_rates_sec)
        elif len(inst_rates) > 0:
            # Fallback to instantaneous rate if global alignment is missing
            mean_rate_sec = np.mean(inst_rates)
            mean_rate_min = mean_rate_sec * 60
            mean_interval = np.mean(ibi)
            breath_count = len(inst_rates)
        
        result = {
            **bout.to_dict(),
            'mean_resp_100hz': sig_mean,
            'min_resp_100hz': sig_min,
            'max_resp_100hz': sig_max,
            'resp_range': sig_range,
            'mean_rate_per_sec': mean_rate_sec,
            'mean_rate_per_min': mean_rate_min,
            'mean_interval_sec': mean_interval,
            'breath_count': breath_count,
            'ibi_list': ibi,                 # NEW: inter-breath intervals
            'inst_rate_list_hz': inst_rates  # NEW: instantaneous rates
        }
        
        results.append(result)
    
    result_df = pd.DataFrame(results)
    print(f"Aligned data for {len(result_df)} behavioral bouts (with IBI/instantaneous rates)")
    
    return result_df
```

```python
import h5py
import numpy as np
import pandas as pd
from scipy import signal
import glob
import os
import matplotlib.pyplot as plt

def load_respiratory_data(h5_file_path):
    with h5py.File(h5_file_path, 'r') as f:
        data = f['resp'][:].flatten()
        print(f"Loaded respiratory data: {len(data)} samples")
        print(f"Data range: {data.min():.1f} to {data.max():.1f}")
    return data

def filter_respiratory_signal(data, original_fs=20000, target_fs=100):
    print(f"Original sampling rate: {original_fs} Hz")
    print(f"Target sampling rate: {target_fs} Hz")
    
    nyquist = original_fs / 2
    cutoff = target_fs / 2
    
    b, a = signal.butter(5, cutoff/nyquist, 'low')
    filtered = signal.filtfilt(b, a, data)
    print(f"Applied filter with cutoff: {cutoff} Hz")
    
    downsample_factor = original_fs // target_fs
    downsampled = filtered[::downsample_factor]
    
    print(f"Downsampled by factor {downsample_factor}: {len(downsampled)} samples at {target_fs} Hz")
    print(f"Duration: {len(downsampled) / target_fs:.2f} seconds")
    
    return downsampled, target_fs

def create_time_vector(signal_data, fs):
    time_vec = np.arange(len(signal_data)) / fs
    print(f"Time vector: 0.00 to {time_vec[-1]:.2f} seconds")
    return time_vec

def detect_breathing_events(signal_data, fs=100):
    normalized = (signal_data - np.mean(signal_data)) / np.std(signal_data)
    
    min_distance = fs // 4
    prominence = 0.5
    height = 0.1
    
    peaks, _ = signal.find_peaks(normalized, distance=min_distance, prominence=prominence, height=height)
    
    print(f"Found {len(peaks)} breathing events")
    
    if len(peaks) < 2:
        print("Warning: Not enough events detected")
        return np.array([]), np.array([]), np.array([])
    
    peak_times = peaks / fs
    intervals = np.diff(peak_times)
    interval_times = peak_times[1:]
    
    print(f"Interval range: {intervals.min():.3f} to {intervals.max():.3f} seconds")
    print(f"Mean interval: {intervals.mean():.3f} seconds")
    
    return peaks, intervals, interval_times

def calculate_breathing_rates(intervals, times):
    rates_per_sec = 1 / intervals
    rates_per_min = rates_per_sec * 60
    
    print(f"Rate range: {rates_per_sec.min():.3f} to {rates_per_sec.max():.3f} breaths/sec")
    print(f"Rate range: {rates_per_min.min():.1f} to {rates_per_min.max():.1f} breaths/min")
    print(f"Mean rate: {rates_per_sec.mean():.3f} breaths/sec ({rates_per_min.mean():.1f} breaths/min)")
    
    return rates_per_sec, rates_per_min

def get_video_fps(filename):
    fps_dict = {
        "RI1_s1_1": 15,
        "RI1_s2_3": 15,
        "RI1_s3_6": 29,
        "RI2_s1_1": 15,
        "RI2_s2_4": 15
    }
    
    for key, fps in fps_dict.items():
        if key in filename:
            return fps
    return 30

def load_behavior_data(excel_path):
    df = pd.read_excel(excel_path)
    print(f"Loaded behavior data: {df.shape}")
    print(f"Columns: {list(df.columns)}")
    
    fps = get_video_fps(os.path.basename(excel_path))
    print(f"Using {fps} FPS for conversion")
    
    df['time_sec'] = df['frame_number'] / fps
    df['fps'] = fps
    
    print(f"Time range: {df['time_sec'].min():.2f} to {df['time_sec'].max():.2f} seconds")
    
    return df, fps

def group_behavior_bouts(df, min_duration=500):
    df['transition'] = (df['syllable'] != df['syllable'].shift()).astype(int)
    df['group'] = df['transition'].cumsum()
    
    bouts = []
    
    for group_id, group in df.groupby('group'):
        syllable = group['syllable'].iloc[0]
        start_frame = group['frame_number'].min()
        end_frame = group['frame_number'].max()
        start_time = group['time_sec'].min()
        end_time = group['time_sec'].max()
        duration = (end_time - start_time) * 1000
        frame_count = len(group)
        fps = group['fps'].iloc[0]
        
        if duration >= min_duration:
            bouts.append({
                'syllable': syllable,
                'start_frame': start_frame,
                'end_frame': end_frame,
                'start_time': start_time,
                'end_time': end_time,
                'duration_ms': duration,
                'frame_count': frame_count,
                'fps': fps,
                'video': group['video_name'].iloc[0] if 'video_name' in group.columns else 'unknown',
                'track': group['track_name'].iloc[0] if 'track_name' in group.columns else 'track0'
            })
    
    bout_df = pd.DataFrame(bouts)
    print(f"Created {len(bout_df)} bouts ≥{min_duration}ms from {len(df)} frames")
    
    if len(bout_df) > 0:
        print(f"Duration range: {bout_df['duration_ms'].min():.0f} - {bout_df['duration_ms'].max():.0f} ms")
    
    return bout_df

def align_respiratory_behavioral_data(bouts, time_vec, signal_data, rates_per_sec, rates_per_min, intervals, interval_times):
    results = []
    
    for idx, bout in bouts.iterrows():
        start = bout['start_time']
        end = bout['end_time']
        
        signal_mask = (time_vec >= start) & (time_vec <= end)
        bout_signal = signal_data[signal_mask]
        
        if len(bout_signal) > 0:
            sig_mean = np.mean(bout_signal)
            sig_min = np.min(bout_signal)
            sig_max = np.max(bout_signal)
            sig_range = sig_max - sig_min
        else:
            sig_mean = sig_min = sig_max = sig_range = np.nan
        
        rate_mask = (interval_times >= start) & (interval_times <= end)
        bout_rates_sec = rates_per_sec[rate_mask]
        bout_rates_min = rates_per_min[rate_mask]
        bout_intervals = intervals[rate_mask]
        
        if len(bout_rates_sec) > 0:
            mean_rate_sec = np.mean(bout_rates_sec)
            mean_rate_min = np.mean(bout_rates_min)
            mean_interval = np.mean(bout_intervals)
            breath_count = len(bout_rates_sec)
        else:
            mid_time = (start + end) / 2
            if len(interval_times) > 0:
                mean_rate_sec = np.interp(mid_time, interval_times, rates_per_sec)
                mean_rate_min = mean_rate_sec * 60
                mean_interval = 1 / mean_rate_sec if mean_rate_sec > 0 else np.nan
            else:
                mean_rate_sec = mean_rate_min = mean_interval = np.nan
            
            breath_count = 0
        
        result = {
            **bout.to_dict(),
            'mean_resp_100hz': sig_mean,
            'min_resp_100hz': sig_min,
            'max_resp_100hz': sig_max,
            'resp_range': sig_range,
            'mean_rate_per_sec': mean_rate_sec,
            'mean_rate_per_min': mean_rate_min,
            'mean_interval_sec': mean_interval,
            'breath_count': breath_count
        }
        
        results.append(result)
    
    result_df = pd.DataFrame(results)
    print(f"Aligned data for {len(result_df)} behavioral bouts")
    
    return result_df

def process_session(h5_file, excel_file, min_duration=500):
    print(f"=== Processing {os.path.basename(h5_file)} with {os.path.basename(excel_file)} ===")
    
    resp_data = load_respiratory_data(h5_file)
    filtered_resp, fs = filter_respiratory_signal(resp_data)
    time_vec = create_time_vector(filtered_resp, fs)
    peaks, intervals, interval_times = detect_breathing_events(filtered_resp, fs)
    
    if len(intervals) > 0:
        rates_sec, rates_min = calculate_breathing_rates(intervals, interval_times)
    else:
        rates_sec = rates_min = np.array([])
    
    behavior_df, fps = load_behavior_data(excel_file)
    bouts = group_behavior_bouts(behavior_df, min_duration)
    
    if len(bouts) > 0 and len(intervals) > 0:
        final_df = align_respiratory_behavioral_data(bouts, time_vec, filtered_resp, rates_sec, rates_min, intervals, interval_times)
    else:
        print("No valid data for alignment")
        return None
    
    h5_name = os.path.basename(h5_file).replace('.h5', '')
    excel_name = os.path.basename(excel_file).replace('.xlsx', '').replace('.xls', '')
    output_file = f"FINAL_PROCESSED_{excel_name}_no_std.xlsx"
    
    final_df.to_excel(output_file, index=False)
    print(f"Saved data: {output_file}")
    
    if len(final_df) > 0:
        valid_rates = final_df['mean_rate_per_min'].dropna()
        if len(valid_rates) > 0:
            print(f"Mean breathing rate: {valid_rates.mean():.1f} breaths/min")
            print(f"Rate range: {valid_rates.min():.1f} - {valid_rates.max():.1f} breaths/min")
    
    return final_df

def run_analysis():
    resp_dir = r"C:\Users\thoma\Code\ResearchCode\KeypointMoSeqPadilla-CoreanoLab\Keypoint RI1+RI2\Resp_h5"
    behavior_dir = r"C:\Users\thoma\Code\ResearchCode\KeypointMoSeqPadilla-CoreanoLab\Keypoint RI1+RI2\Excel_Results"
    
    h5_files = glob.glob(os.path.join(resp_dir, "*.h5"))
    excel_files = glob.glob(os.path.join(behavior_dir, "*track0*.xlsx"))
    
    print(f"Found {len(h5_files)} h5 files and {len(excel_files)} Excel files")
    
    results = {}
    
    for h5_file in h5_files:
        h5_id = os.path.basename(h5_file).replace('.h5', '')
        
        matched_excel = None
        for excel_file in excel_files:
            parts = h5_id.split('_')
            if len(parts) >= 3:
                session_id = '_'.join(parts[:3])
                if session_id in excel_file:
                    matched_excel = excel_file
                    break
        
        if matched_excel:
            try:
                result = process_session(h5_file, matched_excel)
                if result is not None:
                    results[h5_id] = result
            except Exception as e:
                print(f"Error processing {h5_id}: {e}")
        else:
            print(f"No matching file found for {h5_id}")
    
    print(f"\nAnalysis complete")
    print(f"Processed {len(results)} sessions")
    
    return results

def create_single_master_file():
    """
    Creates a single master CSV/Excel file with all sessions combined.
    One row = one bout, with explicit session and source metadata.
    """
    processed_files = glob.glob("FINAL_PROCESSED_*_no_std.xlsx")
    
    print(f"Found {len(processed_files)} processed files")
    
    all_data = []
    
    for file in processed_files:
        df = pd.read_excel(file)
        filename = os.path.basename(file)
        
        # Extract session info
        if "-RI1_" in filename:
            ri_group = "RI1"
            session_part = filename.split("-RI1_")[1]
        elif "-RI2_" in filename:
            ri_group = "RI2"
            session_part = filename.split("-RI2_")[1]
        else:
            print(f"Warning: Could not parse RI group from {filename}")
            continue
        
        # Extract session components (e.g., s1_1, s2_3, etc.)
        session_info = session_part.split("_")[0:3]
        session_id = f"{ri_group}_{'_'.join(session_info)}"
        
        # Find corresponding h5 file name
        source_h5 = f"{session_id}.h5"
        
        # Add metadata columns
        df['session_id'] = session_id
        df['ri_group'] = ri_group
        df['source_h5'] = source_h5
        df['source_behavior_file'] = filename.replace("FINAL_PROCESSED_", "").replace("_no_std.xlsx", ".xlsx")
        
        all_data.append(df)
        print(f"  Loaded {session_id}: {len(df)} bouts")
    
    if not all_data:
        print("No data found to combine!")
        return None
    
    # Combine all dataframes
    master_df = pd.concat(all_data, ignore_index=True)
    
    # Reorder columns to put metadata first
    metadata_cols = ['session_id', 'ri_group', 'source_h5', 'source_behavior_file']
    other_cols = [col for col in master_df.columns if col not in metadata_cols]
    master_df = master_df[metadata_cols + other_cols]
    
    # Save as both CSV and Excel
    csv_file = "MASTER_Respiratory_Behavioral_AllSessions.csv"
    excel_file = "MASTER_Respiratory_Behavioral_AllSessions.xlsx"
    
    master_df.to_csv(csv_file, index=False)
    print(f"\n✅ Saved CSV: {csv_file}")
    print(f"   Total bouts: {len(master_df)}")
    
    master_df.to_excel(excel_file, index=False)
    print(f"✅ Saved Excel: {excel_file}")
    
    # Print summary statistics
    print(f"\n📊 Summary:")
    print(f"   Total sessions: {master_df['session_id'].nunique()}")
    print(f"   RI1 sessions: {len(master_df[master_df['ri_group'] == 'RI1']['session_id'].unique())}")
    print(f"   RI2 sessions: {len(master_df[master_df['ri_group'] == 'RI2']['session_id'].unique())}")
    print(f"   Total bouts: {len(master_df)}")
    print(f"   RI1 bouts: {len(master_df[master_df['ri_group'] == 'RI1'])}")
    print(f"   RI2 bouts: {len(master_df[master_df['ri_group'] == 'RI2'])}")
    
    if 'mean_rate_per_min' in master_df.columns:
        print(f"\n🫁 Breathing rates:")
        for ri_group in ['RI1', 'RI2']:
            group_data = master_df[master_df['ri_group'] == ri_group]['mean_rate_per_min'].dropna()
            if len(group_data) > 0:
                print(f"   {ri_group}: {group_data.mean():.1f} ± {group_data.std():.1f} breaths/min (n={len(group_data)})")
    
    return master_df

results = run_analysis()
master_df = create_single_master_file()
```

```python
#Low pass filter func.
def apply_lowpass_filter(signal, fs, target_rate):
    
    cutoff_hz = min(0.45 * fs, 20.0) if target_rate >= fs else target_rate / 2.0
    Wn = cutoff_hz / (fs / 2.0)
    Wn = min(max(Wn, 0.0001), 0.9999) 
    
    b, a = butter(N=4, Wn=Wn, btype='low')
    filtered = filtfilt(b, a, signal)
    
    print(f"Applied low-pass filter: {cutoff_hz:.2f} Hz cutoff")
    return filtered
```

```python
#Resampling Func.
def resample_signal(signal, duration_sec, target_rate):
 
    N_out = int(round(duration_sec * target_rate))
    resampled = resample(signal, N_out)
    
    print(f"Resampled: {len(signal)} -> {len(resampled)} samples at {target_rate} Hz")
    return resampled
```

```python
#Band-pass filter
def apply_bandpass_filter(signal, sampling_rate, lowcut=0.5, highcut=15.0):
   
    cleaned = nk.signal_filter(
        signal, lowcut=lowcut, highcut=highcut,
        method="butterworth", sampling_rate=sampling_rate, order=2
    )
    
    print(f"Applied band-pass filter: {lowcut}-{highcut} Hz")
    return cleaned
```

```python
import h5py
import numpy as np
from scipy.signal import butter, filtfilt, resample
import neurokit2 as nk
import os

def load_clean_resp_signal(h5_file, target_rate=100):

    try:
        with h5py.File(h5_file, 'r') as f:
            resp = f['resp'][:].flatten()
            ekg_meta = dict(f['ekg_metadata'].attrs)
        duration_sec = float(ekg_meta['duration_sec'])
        fs = len(resp) / duration_sec
        print(f"Loaded {os.path.basename(h5_file)}: {len(resp)} samples, {fs:.2f} Hz, {duration_sec:.2f} sec")
    except Exception as e:
        print(f"Error loading {h5_file}: {e}")
        return None, None, None, None
    
    # Process signal
    filtered = apply_lowpass_filter(resp, fs, target_rate)
    resampled = resample_signal(filtered, duration_sec, target_rate)
    cleaned = apply_bandpass_filter(resampled, target_rate)
    time_vector = np.arange(len(cleaned)) / float(target_rate)
    
    return cleaned, time_vector, float(target_rate), os.path.basename(h5_file)
```

```python
import zipfile
import tempfile

def extract_h5_files_from_zip(zip_path):
    
    temp_dir = tempfile.mkdtemp()
    
    print(f"Extracting {zip_path}...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(temp_dir)
    
    # Find all H5 files
    h5_files = []
    for root, dirs, files in os.walk(temp_dir):
        for file in files:
            if file.endswith('.h5'):
                h5_files.append(os.path.join(root, file))
    
    print(f"Found {len(h5_files)} H5 files")
    return h5_files, temp_dir
```

```python
#Batch processing
def process_multiple_h5_files(h5_files, target_rate=100):
    
    processed_data = {}
    
    for h5_file in h5_files:
        filename = os.path.basename(h5_file)
        print(f"\nProcessing: {filename}")
        
        resp_cleaned, time_vec, fs_out, file_id = load_clean_resp_signal(h5_file, target_rate)
        
        if resp_cleaned is not None:
            processed_data[file_id] = {
                'respiration': resp_cleaned,
                'time': time_vec,
                'sampling_rate': fs_out,
                'duration_sec': len(resp_cleaned) / fs_out,
                'filename': filename
            }
            print(f"Success: {len(resp_cleaned)} samples")
        else:
            print(f"Failed: {filename}")
    
    return processed_data
```

```python
#Save data
import pandas as pd

def save_processed_data(processed_data, output_dir):
  
    output_file = os.path.join(output_dir, "processed_respiration_data.pkl")
    pd.to_pickle(processed_data, output_file)
    print(f"Saved processed data to: {output_file}")
  
    summary_data = []
    for file_id, data in processed_data.items():
        summary_data.append({
            'filename': file_id,
            'duration_sec': data['duration_sec'],
            'sampling_rate': data['sampling_rate'],
            'num_samples': len(data['respiration']),
            'mean_resp': np.mean(data['respiration']),
            'std_resp': np.std(data['respiration'])
        })
    
    summary_df = pd.DataFrame(summary_data)
    summary_file = os.path.join(output_dir, "respiration_summary.xlsx")
    summary_df.to_excel(summary_file, index=False)
    print(f"Saved summary to: {summary_file}")
    
    return summary_df
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import find_peaks
import os

# Load your processed respiration data
processed_resp_data = pd.read_pickle(r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results\processed_respiration_data.pkl")

def calculate_respiratory_metrics(resp_signal, resp_time, sampling_rate=100):
    """
    Calculate detailed respiratory metrics from respiration signal
    """
    if len(resp_signal) < 10:  # Too short to analyze
        return {}
    
    # Find peaks (inspirations) - look for positive peaks
    peak_height = np.std(resp_signal) * 0.3  # Adaptive threshold
    min_distance = int(sampling_rate * 0.5)  # Minimum 0.5 seconds between breaths
    
    peaks, peak_properties = find_peaks(resp_signal, 
                                       height=peak_height, 
                                       distance=min_distance)
    
    # Find troughs (expirations) - look for negative peaks
    troughs, trough_properties = find_peaks(-resp_signal, 
                                          height=peak_height, 
                                          distance=min_distance)
    
    if len(peaks) < 2:  # Need at least 2 breaths
        return {
            'respiratory_rate_per_sec': 0,
            'respiratory_rate_per_min': 0,
            'interbreath_interval_mean': 0,
            'interbreath_interval_std': 0,
            'breath_count': 0,
            'breathing_amplitude_mean': 0,
            'breathing_amplitude_std': 0,
            'inspiration_duration_mean': 0,
            'expiration_duration_mean': 0,
            'respiratory_variability': 0
        }
    
    # Calculate interbreath intervals (peak to peak)
    peak_times = resp_time[peaks]
    interbreath_intervals = np.diff(peak_times)
    
    # Respiratory rate calculations
    total_duration = resp_time[-1] - resp_time[0]
    breath_count = len(peaks)
    respiratory_rate_per_sec = breath_count / total_duration if total_duration > 0 else 0
    respiratory_rate_per_min = respiratory_rate_per_sec * 60
    
    # Breathing amplitudes (peak to trough differences)
    amplitudes = []
    inspiration_durations = []
    expiration_durations = []
    
    for peak_idx in peaks:
        peak_value = resp_signal[peak_idx]
        
        # Find nearest trough before and after peak
        troughs_before = troughs[troughs < peak_idx]
        troughs_after = troughs[troughs > peak_idx]
        
        if len(troughs_before) > 0:
            trough_before_idx = troughs_before[-1]
            trough_before_value = resp_signal[trough_before_idx]
            amplitudes.append(peak_value - trough_before_value)
            
            # Inspiration duration (trough to peak)
            inspiration_duration = (peak_idx - trough_before_idx) / sampling_rate
            inspiration_durations.append(inspiration_duration)
        
        if len(troughs_after) > 0:
            trough_after_idx = troughs_after[0]
            
            # Expiration duration (peak to trough)
            expiration_duration = (trough_after_idx - peak_idx) / sampling_rate
            expiration_durations.append(expiration_duration)
    
    # Calculate metrics
    metrics = {
        'respiratory_rate_per_sec': respiratory_rate_per_sec,
        'respiratory_rate_per_min': respiratory_rate_per_min,
        'interbreath_interval_mean': np.mean(interbreath_intervals) if len(interbreath_intervals) > 0 else 0,
        'interbreath_interval_std': np.std(interbreath_intervals) if len(interbreath_intervals) > 0 else 0,
        'breath_count': breath_count,
        'breathing_amplitude_mean': np.mean(amplitudes) if len(amplitudes) > 0 else 0,
        'breathing_amplitude_std': np.std(amplitudes) if len(amplitudes) > 0 else 0,
        'inspiration_duration_mean': np.mean(inspiration_durations) if len(inspiration_durations) > 0 else 0,
        'expiration_duration_mean': np.mean(expiration_durations) if len(expiration_durations) > 0 else 0,
        'respiratory_variability': np.std(interbreath_intervals) / np.mean(interbreath_intervals) if len(interbreath_intervals) > 0 and np.mean(interbreath_intervals) > 0 else 0
    }
    
    return metrics

def extract_resp_during_bout(resp_data, bout_start_frame, bout_end_frame, video_fps, resp_fps=100):
    """
    Extract respiration data during a specific syllable bout
    """
    # Convert frame numbers to time
    start_time = bout_start_frame / video_fps
    end_time = bout_end_frame / video_fps
    
    # Convert time to respiration indices
    start_idx = int(start_time * resp_fps)
    end_idx = int(end_time * resp_fps)
    
    # Extract respiration during bout
    if start_idx < len(resp_data['respiration']) and end_idx < len(resp_data['respiration']):
        bout_resp = resp_data['respiration'][start_idx:end_idx]
        bout_time = resp_data['time'][start_idx:end_idx] - resp_data['time'][start_idx]  # Start from 0
        return bout_resp, bout_time
    else:
        return None, None

def analyze_resp_during_syllables(bout_details_file, processed_resp_data, min_bout_duration=0.5):
    """
    Analyze respiration patterns during syllable bouts with detailed metrics
    """
    # Load bout data
    bout_df = pd.read_excel(bout_details_file)
    
    # Filter out bouts shorter than 0.5 seconds
    print(f"Original bouts: {len(bout_df)}")
    filtered_bouts = bout_df[bout_df['bout_duration_seconds'] >= min_bout_duration]
    print(f"Filtered bouts (≥{min_bout_duration} seconds): {len(filtered_bouts)}")
    print(f"Removed {len(bout_df) - len(filtered_bouts)} short bouts")
    
    resp_bout_data = []
    
    for _, bout in filtered_bouts.iterrows():
        video_name = bout['video_name']
        
        # Find matching respiration file
        resp_key = f"{video_name}_merged.h5"
        
        if resp_key in processed_resp_data:
            resp_data = processed_resp_data[resp_key]
            video_fps = bout['video_fps']
            
            # Extract respiration during this bout
            bout_resp, bout_time = extract_resp_during_bout(
                resp_data, 
                bout['bout_start_frame'], 
                bout['bout_end_frame'], 
                video_fps
            )
            
            if bout_resp is not None:
                # Calculate detailed respiratory metrics
                resp_metrics = calculate_respiratory_metrics(bout_resp, bout_time)
                
                # Combine bout info with respiratory metrics
                bout_data = {
                    'syllable_id': bout['syllable_id'],
                    'video_name': bout['video_name'],
                    'track_number': bout['track_number'],
                    'bout_start_frame': bout['bout_start_frame'],
                    'bout_end_frame': bout['bout_end_frame'],
                    'bout_duration_sec': bout['bout_duration_seconds'],
                    'video_fps': bout['video_fps'],
                    'resp_mean': np.mean(bout_resp),
                    'resp_std': np.std(bout_resp),
                    'resp_range': np.max(bout_resp) - np.min(bout_resp),
                }
                
                # Add all respiratory metrics
                bout_data.update(resp_metrics)
                
                resp_bout_data.append(bout_data)
    
    return resp_bout_data

def create_detailed_resp_dataframe(resp_bout_data):
    """
    Create detailed DataFrame with all respiratory metrics
    """
    df = pd.DataFrame(resp_bout_data)
    
    # Reorder columns for better readability
    column_order = [
        'syllable_id', 'video_name', 'track_number', 
        'bout_start_frame', 'bout_end_frame', 'bout_duration_sec', 'video_fps',
        'respiratory_rate_per_sec', 'respiratory_rate_per_min', 
        'interbreath_interval_mean', 'interbreath_interval_std',
        'breath_count', 'breathing_amplitude_mean', 'breathing_amplitude_std',
        'inspiration_duration_mean', 'expiration_duration_mean', 
        'respiratory_variability', 'resp_mean', 'resp_std', 'resp_range'
    ]
    
    # Only include columns that exist
    available_columns = [col for col in column_order if col in df.columns]
    df = df[available_columns]
    
    return df

# Execute the analysis
bout_details_file = r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results\detailed_bout_frame_locations_correct_fps_FIXED.xlsx"

print("Analyzing respiration during syllable bouts (≥0.5 seconds)...")
resp_bout_data = analyze_resp_during_syllables(bout_details_file, processed_resp_data, min_bout_duration=0.5)

print(f"Found respiration data for {len(resp_bout_data)} syllable bouts")

# Create detailed DataFrame
detailed_resp_df = create_detailed_resp_dataframe(resp_bout_data)
print("\nFirst few rows of detailed respiration data:")
print(detailed_resp_df.head())

# Save results
output_dir = r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results"

# Save detailed data (like your screenshot)
detailed_file = os.path.join(output_dir, "detailed_respiration_syllable_analysis_filtered.xlsx")
detailed_resp_df.to_excel(detailed_file, index=False)

print(f"\nSaved detailed analysis to: {detailed_file}")
print("Analysis complete!")
```

```python
import numpy as np
import pandas as pd
from scipy.signal import find_peaks
import glob
import os

def process_respiratory_bouts_with_bandpass_corrected():
    processed_resp_data = pd.read_pickle(r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results\processed_respiration_data_0.5Hz.pkl")
    
    excel_path = r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results"
    excel_files = glob.glob(os.path.join(excel_path, "*track0*.xlsx"))
    
    results = {}
    
    for file_id, resp_data in processed_resp_data.items():
        print(f"Processing {file_id}")
        
        filtered_signal = resp_data['respiration']
        fs = 100
        time_vector = np.arange(len(filtered_signal)) / fs
        
        normalized_signal = (filtered_signal - np.mean(filtered_signal)) / np.std(filtered_signal)
        peaks, _ = find_peaks(normalized_signal, distance=fs//4, prominence=0.5, height=0.1)
        
        print(f"Found {len(peaks)} breathing peaks")
        
        if len(peaks) < 2:
            continue
            
        peak_times = peaks / fs
        ibi_seconds = np.diff(peak_times)
        breathing_rate_1_over_ibi = 1 / ibi_seconds
        breathing_rate_bpm = breathing_rate_1_over_ibi * 60
        breath_times = peak_times[1:]
        
        print(f"IBI range: {ibi_seconds.min():.3f} - {ibi_seconds.max():.3f} seconds")
        print(f"Mean IBI: {ibi_seconds.mean():.3f} seconds")
        
        matching_excel = None
        for excel_file in excel_files:
            if any(session in file_id and session in excel_file for session in 
                   ['RI1_s1_1', 'RI1_s1_2', 'RI1_s2_3', 'RI1_s2_4', 'RI1_s3_5', 'RI1_s3_6', 
                    'RI1_s4_7', 'RI1_s4_8', 'RI1_s5_13', 'RI2_s1_1', 'RI2_s1_2', 'RI2_s2_3', 
                    'RI2_s2_4', 'RI2_s3_5', 'RI2_s3_6', 'RI2_s4_7', 'RI2_s4_8', 'RI2_s5_13']):
                matching_excel = excel_file
                break
        
        if matching_excel:
            print(f"Matched with: {os.path.basename(matching_excel)}")
            behavior_df = pd.read_excel(matching_excel)
            
            fps_map = {"RI1_s1_1": 15, "RI1_s2_3": 15, "RI1_s3_6": 29, "RI2_s1_1": 15, "RI2_s2_4": 15}
            video_fps = 30
            for pattern, fps in fps_map.items():
                if pattern in os.path.basename(matching_excel):
                    video_fps = fps
                    break
            
            print(f"Using {video_fps} FPS")
            
            behavior_df['time_seconds'] = behavior_df['frame_number'] / video_fps
            behavior_df['behavior_transition'] = (behavior_df['syllable'] != behavior_df['syllable'].shift()).astype(int)
            behavior_df['episode_group'] = behavior_df['behavior_transition'].cumsum()
            
            syllable_bouts = []
            
            for group_id, episode_group in behavior_df.groupby('episode_group'):
                syllable_id = episode_group['syllable'].iloc[0]
                
                bout_start_frame = episode_group['frame_number'].min()
                bout_end_frame = episode_group['frame_number'].max()
                bout_start_time_sec = episode_group['time_seconds'].min()
                bout_end_time_sec = episode_group['time_seconds'].max()
                bout_duration_ms = (bout_end_time_sec - bout_start_time_sec) * 1000
                bout_frame_count = len(episode_group)
                
                if bout_duration_ms >= 500:
                    
                    time_mask = (time_vector >= bout_start_time_sec) & (time_vector <= bout_end_time_sec)
                    bout_resp_signal = filtered_signal[time_mask]
                    
                    if len(bout_resp_signal) > 0:
                        mean_resp = np.mean(bout_resp_signal)
                        min_resp = np.min(bout_resp_signal)
                        max_resp = np.max(bout_resp_signal)
                        resp_range = max_resp - min_resp
                    else:
                        mean_resp = min_resp = max_resp = resp_range = np.nan
                    
                    breath_mask = (breath_times >= bout_start_time_sec) & (breath_times <= bout_end_time_sec)
                    bout_1_over_ibi = breathing_rate_1_over_ibi[breath_mask]
                    bout_bpm = breathing_rate_bpm[breath_mask]
                    bout_ibi = ibi_seconds[breath_mask]
                    
                    if len(bout_1_over_ibi) > 0:
                        mean_1_over_ibi = np.mean(bout_1_over_ibi)
                        mean_bpm = np.mean(bout_bpm)
                        mean_ibi = np.mean(bout_ibi)
                        breath_count = len(bout_1_over_ibi)
                    else:
                        mid_time = (bout_start_time_sec + bout_end_time_sec) / 2
                        if len(breath_times) > 0:
                            mean_1_over_ibi = np.interp(mid_time, breath_times, breathing_rate_1_over_ibi)
                            mean_bpm = mean_1_over_ibi * 60
                            mean_ibi = 1 / mean_1_over_ibi if mean_1_over_ibi > 0 else np.nan
                        else:
                            mean_1_over_ibi = mean_bpm = mean_ibi = np.nan
                        breath_count = 0
                    
                    syllable_bouts.append({
                        'syllable_id': syllable_id,
                        'bout_start_frame': bout_start_frame,
                        'bout_end_frame': bout_end_frame,
                        'bout_start_time_sec': bout_start_time_sec,
                        'bout_end_time_sec': bout_end_time_sec,
                        'bout_duration_ms': bout_duration_ms,
                        'bout_frame_count': bout_frame_count,
                        'fps_used': video_fps,
                        'mean_resp_signal_bandpass': mean_resp,
                        'min_resp_signal_bandpass': min_resp,
                        'max_resp_signal_bandpass': max_resp,
                        'resp_signal_range': resp_range,
                        'mean_1_over_IBI': mean_1_over_ibi,
                        'mean_breathing_rate_BPM': mean_bpm,
                        'mean_IBI_seconds': mean_ibi,
                        'breath_count': breath_count,
                        'video_name': episode_group['video_name'].iloc[0] if 'video_name' in episode_group.columns else 'unknown',
                        'track_name': episode_group['track_name'].iloc[0] if 'track_name' in episode_group.columns else 'track0'
                    })
            
            bouts_df = pd.DataFrame(syllable_bouts)
            bouts_df = bouts_df.sort_values('bout_start_time_sec').reset_index(drop=True)
            
            if len(bouts_df) > 0:
                excel_basename = os.path.basename(matching_excel).replace('.xlsx', '')
                output_file = f"FINAL_BOUTS_{excel_basename}.xlsx"
                bouts_df.to_excel(output_file, index=False)
                print(f"Saved: {output_file} with {len(bouts_df)} bouts")
                
                results[file_id] = bouts_df
        else:
            print(f"No matching file found")
    
    return results

def create_master_bout_files():
    bout_files = glob.glob("FINAL_BOUTS_*.xlsx")
    
    ri1_files = [f for f in bout_files if "-RI1_" in f]
    ri2_files = [f for f in bout_files if "-RI2_" in f]
    
    def create_master_file(files, output_name):
        with pd.ExcelWriter(output_name, engine='openpyxl') as writer:
            for file in files:
                df = pd.read_excel(file)
                
                filename = os.path.basename(file)
                if "-RI1_" in filename:
                    session_part = filename.split("-RI1_")[1]
                    sheet_prefix = "RI1_"
                elif "-RI2_" in filename:
                    session_part = filename.split("-RI2_")[1]
                    sheet_prefix = "RI2_"
                
                session_info = session_part.split("_")[0:3]
                sheet_name = f"{sheet_prefix}{'_'.join(session_info)}_track0"
                
                if len(sheet_name) > 31:
                    sheet_name = sheet_name[:31]
                
                df.to_excel(writer, sheet_name=sheet_name, index=False)
                print(f"Added {sheet_name}: {len(df)} bouts")
    
    if ri1_files:
        create_master_file(ri1_files, "RI1_MASTER_Bouts.xlsx")
    if ri2_files:
        create_master_file(ri2_files, "RI2_MASTER_Bouts.xlsx")

print("Processing respiratory bout data")
corrected_bout_results = process_respiratory_bouts_with_bandpass_corrected()
create_master_bout_files()
print(f"Done - processed {len(corrected_bout_results)} files")

if corrected_bout_results:
    sample_key = list(corrected_bout_results.keys())[0]
    sample_data = corrected_bout_results[sample_key]
    
    valid_ibi = sample_data['mean_IBI_seconds'].dropna()
    if len(valid_ibi) > 0:
        print(f"IBI range: {valid_ibi.min():.3f} - {valid_ibi.max():.3f} seconds")
        print(f"BPM range: {valid_ibi.min()*60:.1f} - {valid_ibi.max()*60:.1f}")
```

```python
import numpy as np
import pandas as pd
from scipy.signal import find_peaks
import glob
import os

def improved_peak_detection(signal, fs=100):
    """Improved peak detection for respiratory signals"""
    
    # Calculate signal characteristics
    signal_std = np.std(signal)
    signal_mean = np.mean(signal)
    
    # Use adaptive thresholds
    height_threshold = signal_mean + 0.3 * signal_std
    prominence_threshold = signal_std * 0.4
    
    # Detect peaks with better parameters
    peaks, properties = find_peaks(signal, 
                                  distance=fs * 0.15,  # 0.15s minimum (400 BPM max)
                                  height=height_threshold,
                                  prominence=prominence_threshold,
                                  width=3)  # Minimum width requirement
    
    return peaks

def create_improved_syllable_bout_files():
    processed_resp_data = pd.read_pickle(r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results\processed_respiration_data_0.5Hz.pkl")
    
    excel_path = r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results"
    excel_files = glob.glob(os.path.join(excel_path, "*track0*.xlsx"))
    
    results = {}
    
    for file_id, resp_data in processed_resp_data.items():
        print(f"Processing {file_id}")
        
        filtered_signal = resp_data['respiration']
        fs = 100
        time_vector = np.arange(len(filtered_signal)) / fs
        
        # Use improved peak detection
        peaks = improved_peak_detection(filtered_signal, fs)
        
        print(f"Found {len(peaks)} breathing peaks (improved method)")
        
        if len(peaks) < 2:
            continue
            
        peak_times = peaks / fs
        ibi_seconds = np.diff(peak_times)
        breathing_rate_1_over_ibi = 1 / ibi_seconds
        breathing_rate_bpm = breathing_rate_1_over_ibi * 60
        breath_times = peak_times[1:]
        
        print(f"Breathing rate: {breathing_rate_bpm.mean():.1f} BPM (range: {breathing_rate_bpm.min():.1f}-{breathing_rate_bpm.max():.1f})")
        
        matching_excel = None
        for excel_file in excel_files:
            if any(session in file_id and session in excel_file for session in 
                   ['RI1_s1_1', 'RI1_s1_2', 'RI1_s2_3', 'RI1_s2_4', 'RI1_s3_5', 'RI1_s3_6', 
                    'RI1_s4_7', 'RI1_s4_8', 'RI1_s5_13', 'RI2_s1_1', 'RI2_s1_2', 'RI2_s2_3', 
                    'RI2_s2_4', 'RI2_s3_5', 'RI2_s3_6', 'RI2_s4_7', 'RI2_s4_8', 'RI2_s5_13']):
                matching_excel = excel_file
                break
        
        if matching_excel:
            print(f"Matched with: {os.path.basename(matching_excel)}")
            behavior_df = pd.read_excel(matching_excel)
            
            fps_map = {"RI1_s1_1": 15, "RI1_s2_3": 15, "RI1_s3_6": 29, "RI2_s1_1": 15, "RI2_s2_4": 15}
            video_fps = 30
            for pattern, fps in fps_map.items():
                if pattern in os.path.basename(matching_excel):
                    video_fps = fps
                    break
            
            behavior_df['time_seconds'] = behavior_df['frame_number'] / video_fps
            behavior_df['behavior_transition'] = (behavior_df['syllable'] != behavior_df['syllable'].shift()).astype(int)
            behavior_df['episode_group'] = behavior_df['behavior_transition'].cumsum()
            
            syllable_bouts = []
            
            for group_id, episode_group in behavior_df.groupby('episode_group'):
                syllable_id = episode_group['syllable'].iloc[0]
                
                bout_start_frame = episode_group['frame_number'].min()
                bout_end_frame = episode_group['frame_number'].max()
                bout_start_time_sec = episode_group['time_seconds'].min()
                bout_end_time_sec = episode_group['time_seconds'].max()
                bout_duration_ms = (bout_end_time_sec - bout_start_time_sec) * 1000
                bout_frame_count = len(episode_group)
                
                if bout_duration_ms >= 500:
                    
                    time_mask = (time_vector >= bout_start_time_sec) & (time_vector <= bout_end_time_sec)
                    bout_resp_signal = filtered_signal[time_mask]
                    
                    if len(bout_resp_signal) > 0:
                        mean_resp = np.mean(bout_resp_signal)
                        min_resp = np.min(bout_resp_signal)
                        max_resp = np.max(bout_resp_signal)
                        resp_range = max_resp - min_resp
                    else:
                        mean_resp = min_resp = max_resp = resp_range = np.nan
                    
                    breath_mask = (breath_times >= bout_start_time_sec) & (breath_times <= bout_end_time_sec)
                    bout_1_over_ibi = breathing_rate_1_over_ibi[breath_mask]
                    bout_bpm = breathing_rate_bpm[breath_mask]
                    bout_ibi = ibi_seconds[breath_mask]
                    
                    if len(bout_1_over_ibi) > 0:
                        mean_1_over_ibi = np.mean(bout_1_over_ibi)
                        mean_bpm = np.mean(bout_bpm)
                        mean_ibi = np.mean(bout_ibi)
                        breath_count = len(bout_1_over_ibi)
                    else:
                        mid_time = (bout_start_time_sec + bout_end_time_sec) / 2
                        if len(breath_times) > 0:
                            mean_1_over_ibi = np.interp(mid_time, breath_times, breathing_rate_1_over_ibi)
                            mean_bpm = mean_1_over_ibi * 60
                            mean_ibi = 1 / mean_1_over_ibi if mean_1_over_ibi > 0 else np.nan
                        else:
                            mean_1_over_ibi = mean_bpm = mean_ibi = np.nan
                        breath_count = 0
                    
                    syllable_bouts.append({
                        'syllable_id': syllable_id,
                        'bout_start_frame': bout_start_frame,
                        'bout_end_frame': bout_end_frame,
                        'bout_start_time_sec': bout_start_time_sec,
                        'bout_end_time_sec': bout_end_time_sec,
                        'bout_duration_ms': bout_duration_ms,
                        'bout_frame_count': bout_frame_count,
                        'fps_used': video_fps,
                        'mean_resp_signal_bandpass': mean_resp,
                        'min_resp_signal_bandpass': min_resp,
                        'max_resp_signal_bandpass': max_resp,
                        'resp_signal_range': resp_range,
                        'mean_1_over_IBI': mean_1_over_ibi,
                        'mean_breathing_rate_BPM': mean_bpm,
                        'mean_IBI_seconds': mean_ibi,
                        'breath_count': breath_count,
                        'video_name': episode_group['video_name'].iloc[0] if 'video_name' in episode_group.columns else 'unknown',
                        'track_name': episode_group['track_name'].iloc[0] if 'track_name' in episode_group.columns else 'track0'
                    })
            
            bouts_df = pd.DataFrame(syllable_bouts)
            bouts_df = bouts_df.sort_values('bout_start_time_sec').reset_index(drop=True)
            
            if len(bouts_df) > 0:
                excel_basename = os.path.basename(matching_excel).replace('.xlsx', '')
                output_file = f"FINAL_IMPROVED_{excel_basename}.xlsx"
                bouts_df.to_excel(output_file, index=False)
                print(f"Saved: {output_file} with {len(bouts_df)} bouts")
                
                results[file_id] = bouts_df
    
    return results

def create_final_master_files():
    bout_files = glob.glob("FINAL_IMPROVED_*.xlsx")
    
    ri1_files = [f for f in bout_files if "-RI1_" in f]
    ri2_files = [f for f in bout_files if "-RI2_" in f]
    
    print(f"Found {len(ri1_files)} RI1 files and {len(ri2_files)} RI2 files for master files")
    
    def create_master_file(files, output_name):
        with pd.ExcelWriter(output_name, engine='openpyxl') as writer:
            for file in files:
                df = pd.read_excel(file)
                
                filename = os.path.basename(file)
                if "-RI1_" in filename:
                    session_part = filename.split("-RI1_")[1]
                    sheet_prefix = "RI1_"
                elif "-RI2_" in filename:
                    session_part = filename.split("-RI2_")[1]
                    sheet_prefix = "RI2_"
                
                session_info = session_part.split("_")[0:3]
                sheet_name = f"{sheet_prefix}{'_'.join(session_info)}_track0"
                
                if len(sheet_name) > 31:
                    sheet_name = sheet_name[:31]
                
                df.to_excel(writer, sheet_name=sheet_name, index=False)
                print(f"Added {sheet_name}: {len(df)} bouts")
    
    if ri1_files:
        create_master_file(ri1_files, "RI1_MASTER_Final_Improved_Peaks.xlsx")
    if ri2_files:
        create_master_file(ri2_files, "RI2_MASTER_Final_Improved_Peaks.xlsx")

print("Reprocessing with improved peak detection")
improved_results = create_improved_syllable_bout_files()
create_final_master_files()

print(f"Completed - processed {len(improved_results)} files with improved peak detection")
print("Created final master files:")
print("- RI1_MASTER_Final_Improved_Peaks.xlsx") 
print("- RI2_MASTER_Final_Improved_Peaks.xlsx")

# Verify the improvement
if improved_results:
    sample_key = list(improved_results.keys())[0]
    sample_data = improved_results[sample_key]
    
    valid_bpm = sample_data['mean_breathing_rate_BPM'].dropna()
    if len(valid_bpm) > 0:
        print(f"\nImproved results verification:")
        print(f"BPM range: {valid_bpm.min():.1f} - {valid_bpm.max():.1f}")
        print(f"Mean BPM: {valid_bpm.mean():.1f}")
        print(f"This should be more realistic than 480+ BPM!")
```

```python
import h5py
import numpy as np
import pandas as pd
from scipy.signal import find_peaks, butter, filtfilt
import glob
import os
import matplotlib.pyplot as plt
from datetime import datetime

def H(title):
    line = "=" * max(12, len(title))
    print("\n" + line)
    print(title)
    print(line)

H("Imports and runtime state")
print("Libraries loaded")
print(f"Started: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")

H("Global parameters and core assumptions")
RESP_H5_PATH = r"C:\Users\Padilla-Coreano\Documents\tanish\Resp_h5"
EXCEL_PATH = r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results"
PROCESSED_DATA_PATH = r"C:\Users\Padilla-Coreano\Documents\tanis...oint RI1+RI2\Excel_Results\processed_respiration_data_0.5Hz.pkl"

ORIGINAL_SAMPLING_RATE = 20000
TARGET_SAMPLING_RATE = 100
MIN_SYLLABLE_DURATION = 500

PEAK_MIN_DISTANCE = 0.08
PEAK_PROMINENCE_FACTOR = 0.2

FPS_MAPPING = {
    "RI1_s1_1": 15,
    "RI1_s2_3": 15,
    "RI1_s3_6": 29,
    "RI2_s1_1": 15,
    "RI2_s2_4": 15
}
DEFAULT_FPS = 30

print({
    "ORIGINAL_SAMPLING_RATE": ORIGINAL_SAMPLING_RATE,
    "TARGET_SAMPLING_RATE": TARGET_SAMPLING_RATE,
    "MIN_SYLLABLE_DURATION_MS": MIN_SYLLABLE_DURATION,
    "PEAK_MIN_DISTANCE_SEC": PEAK_MIN_DISTANCE,
    "PEAK_PROMINENCE_FACTOR": PEAK_PROMINENCE_FACTOR,
    "DEFAULT_FPS": DEFAULT_FPS,
    "FPS_MAPPING_KEYS": sorted(list(FPS_MAPPING.keys()))
})

H("H5 ingestion primitives")
def list_h5_datasets(h5_path):
    keys = []
    with h5py.File(h5_path, "r") as f:
        def visitor(name, obj):
            if isinstance(obj, h5py.Dataset):
                keys.append(name)
        f.visititems(visitor)
    return sorted(keys)

def read_h5_signal(h5_path, dataset_key):
    with h5py.File(h5_path, "r") as f:
        x = np.array(f[dataset_key])
    return x

print("H5 dataset listing and reading ready")

H("Filtering and resampling primitives")
def butter_lowpass(cutoff_hz, fs_hz, order=4):
    nyq = 0.5 * fs_hz
    normal = cutoff_hz / nyq
    b, a = butter(order, normal, btype="low", analog=False)
    return b, a

def apply_lowpass(x, cutoff_hz, fs_hz, order=4):
    b, a = butter_lowpass(cutoff_hz, fs_hz, order=order)
    return filtfilt(b, a, x)

def resample_by_decimation(x, original_fs, target_fs):
    factor = int(round(original_fs / target_fs))
    if factor <= 0:
        raise ValueError("Invalid decimation factor")
    if original_fs % target_fs != 0:
        raise ValueError("original_fs must be an integer multiple of target_fs for this resampling method")
    return x[::factor]

print("Filter + decimation resampling ready")

H("Normalization primitives (if used before peak detection)")
def zscore(x):
    s = np.std(x)
    m = np.mean(x)
    if s == 0:
        return x * 0
    return (x - m) / s

def center(x):
    return x - np.mean(x)

print("Normalization helpers ready")

H("Peak detection logic")
def detect_breathing_peaks(signal, fs=TARGET_SAMPLING_RATE):
    signal_std = np.std(signal)
    signal_mean = np.mean(signal)

    peaks, properties = find_peaks(
        signal,
        distance=int(fs * PEAK_MIN_DISTANCE),
        prominence=signal_std * PEAK_PROMINENCE_FACTOR,
        height=signal_mean
    )

    return peaks

print("Peak detection ready")

H("Video FPS resolution")
def get_video_fps(filename):
    for pattern, fps in FPS_MAPPING.items():
        if pattern in filename:
            return fps
    return DEFAULT_FPS

print("FPS resolver ready")

H("Respiratory rate and IBI math from peaks")
def calculate_respiratory_metrics(peaks, fs=TARGET_SAMPLING_RATE):
    if len(peaks) < 2:
        return np.array([]), np.array([]), np.array([])

    peak_times = peaks / fs
    ibi_seconds = np.diff(peak_times)
    breathing_rate_1_over_ibi = 1 / ibi_seconds
    breath_times = peak_times[1:]

    return breathing_rate_1_over_ibi, breath_times, ibi_seconds

print("Resp metrics ready")

H("Behavior dataframe schema normalization")
def canonicalize_behavior_columns(df):
    df = df.copy()

    renames = {}
    if "frame_number" not in df.columns:
        if "Frame" in df.columns:
            renames["Frame"] = "frame_number"
        if "frame" in df.columns:
            renames["frame"] = "frame_number"

    if "syllable" not in df.columns:
        if "Syllable" in df.columns:
            renames["Syllable"] = "syllable"
        if "syllable_id" in df.columns:
            renames["syllable_id"] = "syllable"

    if renames:
        df = df.rename(columns=renames)

    if "frame_number" not in df.columns or "syllable" not in df.columns:
        raise ValueError(f"Required columns not found after canonicalization. Columns={list(df.columns)}")

    df = df.dropna(subset=["frame_number", "syllable"])
    df["frame_number"] = df["frame_number"].astype(int)
    df["syllable"] = df["syllable"].astype(int)

    return df

print("Behavior schema normalizer ready")

H("Keypoint-MoSeq syllable bout construction (grouping + duration filtering)")
def create_syllable_bouts(behavior_df, video_fps, min_duration_ms=MIN_SYLLABLE_DURATION):
    behavior_df = behavior_df.copy()
    behavior_df["time_seconds"] = behavior_df["frame_number"] / video_fps
    behavior_df["transition"] = (behavior_df["syllable"] != behavior_df["syllable"].shift()).astype(int)
    behavior_df["bout_group"] = behavior_df["transition"].cumsum()

    bouts_list = []

    for group_id, group in behavior_df.groupby("bout_group"):
        syllable_id = int(group["syllable"].iloc[0])

        start_time = float(group["time_seconds"].min())
        end_time = float(group["time_seconds"].max())
        duration_ms = float((end_time - start_time) * 1000)

        if duration_ms >= min_duration_ms:
            bout_info = {
                "syllable_id": syllable_id,
                "start_frame": int(group["frame_number"].min()),
                "end_frame": int(group["frame_number"].max()),
                "start_time_sec": start_time,
                "end_time_sec": end_time,
                "duration_ms": duration_ms,
                "frame_count": int(len(group)),
                "fps_used": float(video_fps),
            }
            bouts_list.append(bout_info)

    return bouts_list

print("Bout constructor ready")

H("Session key reconciliation and Excel selection")
def find_resp_key(resp_dict, session_pattern):
    matches = [k for k in resp_dict.keys() if session_pattern in k]
    if len(matches) == 0:
        return None
    if len(matches) == 1:
        return matches[0]
    matches_sorted = sorted(matches, key=len)
    return matches_sorted[0]

def pick_excel_file(session_id, excel_files):
    matches = [f for f in excel_files if session_id in os.path.basename(f)]
    if len(matches) == 0:
        return None
    if len(matches) == 1:
        return matches[0]
    matches_sorted = sorted(matches)
    return matches_sorted[0]

print("Key matching helpers ready")

H("Windowing math and per-bout respiratory summaries")
def count_peaks_in_window(peak_times, start_time, end_time, inclusive=True):
    if inclusive:
        mask = (peak_times >= start_time) & (peak_times <= end_time)
    else:
        mask = (peak_times > start_time) & (peak_times < end_time)
    return int(mask.sum())

def summarize_bout_respiration(peak_times, breath_times, breathing_rates_hz, start_time, end_time):
    peak_count = count_peaks_in_window(peak_times, start_time, end_time, inclusive=True)

    mid_time = (start_time + end_time) / 2
    if len(breath_times) > 0 and len(breathing_rates_hz) > 0:
        rate_hz = float(np.interp(mid_time, breath_times, breathing_rates_hz))
        bpm = float(rate_hz * 60)
        ibi = float(1 / rate_hz) if rate_hz > 0 else np.nan
    else:
        rate_hz = np.nan
        bpm = np.nan
        ibi = np.nan

    return {
        "peak_count": peak_count,
        "mean_breathing_rate_hz": rate_hz,
        "mean_bpm": bpm,
        "mean_ibi_sec": ibi
    }

print("Bout respiration summary ready")

H("Single-session integration (peaks + bouts + per-bout features)")
def process_single_session(session_id, respiratory_signal, excel_files):
    print(f"Processing: {session_id}")

    signal = respiratory_signal["respiration"]
    fs = respiratory_signal.get("fs", TARGET_SAMPLING_RATE)

    peaks = detect_breathing_peaks(signal, fs)
    print(f"  Peaks detected: {len(peaks)}")

    if len(peaks) < 2:
        print("  Insufficient peaks - skipping")
        return None

    breathing_rates_hz, breath_times, ibi_values = calculate_respiratory_metrics(peaks, fs)
    peak_times = peaks / fs

    excel_file = pick_excel_file(session_id, excel_files)
    if excel_file is None:
        print(f"  No Excel file found for {session_id}")
        return None

    print(f"  Using Excel file: {os.path.basename(excel_file)}")
    behavior_df = pd.read_excel(excel_file)
    behavior_df = canonicalize_behavior_columns(behavior_df)

    video_fps = get_video_fps(os.path.basename(excel_file))
    bouts_list = create_syllable_bouts(behavior_df, video_fps, MIN_SYLLABLE_DURATION)

    print(f"  Found {len(bouts_list)} syllable bouts")

    processed_bouts = []
    for bout in bouts_list:
        start_time = bout["start_time_sec"]
        end_time = bout["end_time_sec"]
        resp_summary = summarize_bout_respiration(peak_times, breath_times, breathing_rates_hz, start_time, end_time)

        out_row = {
            "syllable_id": bout["syllable_id"],
            "start_time_sec": start_time,
            "end_time_sec": end_time,
            "duration_ms": bout["duration_ms"],
            "frame_count": bout["frame_count"],
            "fps_used": bout["fps_used"],
        }
        out_row.update(resp_summary)
        processed_bouts.append(out_row)

    results_df = pd.DataFrame(processed_bouts)
    results_df = results_df.sort_values("start_time_sec").reset_index(drop=True)

    print(f"  Processed {len(results_df)} syllable bouts")
    return results_df

print("Single-session runner ready")

H("Batch runner and export logic")
def run_complete_analysis():
    print("Starting analysis...")

    respiratory_data = pd.read_pickle(PROCESSED_DATA_PATH)
    excel_files = glob.glob(os.path.join(EXCEL_PATH, "*track0*.xlsx"))

    print(f"Found {len(respiratory_data)} respiratory sessions")
    print(f"Found {len(excel_files)} behavioral files")

    all_results = {}
    successful_sessions = 0

    for session_id, resp_data in respiratory_data.items():
        try:
            result = process_single_session(session_id, resp_data, excel_files)

            if result is not None:
                session_name = session_id.replace(" ", "_").replace("/", "_")
                output_file = f"CORRECTED_PEAK_COUNT_{session_name}_analysis.xlsx"
                result.to_excel(output_file, index=False)
                print(f"  Saved: {output_file}")

                all_results[session_id] = result
                successful_sessions += 1

        except Exception as e:
            print(f"  Error processing {session_id}: {str(e)}")

    print(f"Analysis complete: {successful_sessions}/{len(respiratory_data)} sessions processed")
    return all_results

print("Batch runner ready")

H("Peak-count verification against exported analysis output")
def verify_peak_detection(session_id="RI1_s1_1", test_rows=[2, 5, 8]):
    excel_pattern = f"CORRECTED_PEAK_COUNT_*{session_id}*_analysis.xlsx"
    excel_files = glob.glob(excel_pattern)

    if not excel_files:
        print(f"No Excel file found for {session_id}")
        return None

    excel_file = excel_files[0]
    df = pd.read_excel(excel_file)

    respiratory_data = pd.read_pickle(PROCESSED_DATA_PATH)
    resp_key = find_resp_key(respiratory_data, session_id)
    if resp_key is None:
        print(f"No respiration key found for {session_id}")
        return None

    signal = respiratory_data[resp_key]["respiration"]
    fs = respiratory_data[resp_key].get("fs", TARGET_SAMPLING_RATE)

    peaks = detect_breathing_peaks(signal, fs)
    peak_times = peaks / fs

    matches = 0
    total_tests = len(test_rows)

    plt.figure(figsize=(15, 8))

    for i, row_idx in enumerate(test_rows):
        if row_idx >= len(df):
            continue

        bout = df.iloc[row_idx]
        start_time = float(bout["start_time_sec"])
        end_time = float(bout["end_time_sec"])

        window_mask = (peak_times >= start_time) & (peak_times <= end_time)
        window_peak_times = peak_times[window_mask]

        graph_count = len(window_peak_times)
        excel_count = int(bout["peak_count"])

        if graph_count == excel_count:
            matches += 1
            match_text = "MATCH"
        else:
            match_text = "MISMATCH"

        plt.subplot(len(test_rows), 1, i + 1)

        t = np.arange(len(signal)) / fs
        time_window = (t >= start_time - 0.5) & (t <= end_time + 0.5)
        window_time = t[time_window]
        window_signal = signal[time_window]

        plt.plot(window_time, window_signal)

        for pt in window_peak_times:
            plt.axvline(x=pt, linestyle="--", alpha=0.7)

        plt.axvline(x=start_time, linewidth=2)
        plt.axvline(x=end_time, linewidth=2)

        plt.title(
            f"Row {row_idx}: Excel={excel_count}, Graph={graph_count} {match_text} "
            f"(Syllable {int(bout['syllable_id'])}, {start_time:.2f}-{end_time:.2f}s)"
        )

    plt.suptitle(f"{session_id} Verification - {matches}/{total_tests} matches")
    plt.tight_layout()
    plt.show()

    print(f"Verification: {matches}/{total_tests} syllables match")
    return matches == total_tests

print("Verification routine ready")
```
