```python
# STEP 0: Install required packages if needed
# !pip install keypoint-moseq jax==0.4.26 jaxlib==0.4.26

import keypoint_moseq as kpms
import os
import glob
import numpy as np
from collections import Counter
import matplotlib.pyplot as plt

# STEP 1: Set paths
project_dir = r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2"
sleap_dir = project_dir  # All SLEAP .h5 files should be in this folder

# STEP 2: Gather all SLEAP .h5 files
sleap_files = glob.glob(os.path.join(sleap_dir, "*.h5"))
print(f"Found {len(sleap_files)} SLEAP files.")

if len(sleap_files) == 0:
    raise FileNotFoundError("No .h5 files found in the given directory!")

# STEP 3: Setup project using just the first file to get skeleton/bodypart info
kpms.setup_project(project_dir, sleap_file=sleap_files[0], overwrite=True)

# STEP 4: Configure project
kpms.update_config(
    project_dir,
    video_dir=sleap_dir,
    anterior_bodyparts=["nose"],
    posterior_bodyparts=["tail_base"],
    use_bodyparts=["nose", "right_ear", "left_ear", "middle", "head", "right_mid", "left_mid", "tail_base"],
    fps=30
)
config = lambda: kpms.load_config(project_dir)

# STEP 5: Load keypoints from all videos
coordinates, confidences, bodyparts = kpms.load_keypoints(sleap_files, "sleap")

# STEP 6: Noise calibration
kpms.noise_calibration(project_dir, coordinates, confidences, **config())
plt.close('all')

# STEP 7: PCA (dimensionality reduction)
data, metadata = kpms.format_data(coordinates, confidences, **config())
pca = kpms.fit_pca(**data, **config())
kpms.save_pca(pca, project_dir)
kpms.print_dims_to_explain_variance(pca, 0.9)
kpms.plot_scree(pca, project_dir=project_dir)
kpms.plot_pcs(pca, project_dir=project_dir, **config())

# STEP 8: Initialize and fit model (AR-only)
kpms.update_config(project_dir, latent_dim=4)
model = kpms.init_model(data, pca=pca, **config())
model = kpms.update_hypparams(model, kappa=5000)

model, model_name = kpms.fit_model(
    model,
    data,
    metadata,
    project_dir,
    ar_only=True,
    num_iters=50
)

# STEP 9: Fit full ARHMM model
model, data, metadata, current_iter = kpms.load_checkpoint(project_dir, model_name, iteration=50)
model = kpms.update_hypparams(model, kappa=1e4)
model = kpms.fit_model(
    model,
    data,
    metadata,
    project_dir,
    model_name,
    ar_only=False,
    start_iter=current_iter,
    num_iters=current_iter + 500
)[0]

# STEP 10: Reindex syllables so most frequent = lowest ID
kpms.reindex_syllables_in_checkpoint(project_dir, model_name)

# STEP 11: Load trained model and extract syllables
model, data, metadata, current_iter = kpms.load_checkpoint(project_dir, model_name)
results = kpms.extract_results(model, metadata, project_dir, model_name)

# STEP 12: Get top 20 most common syllables
all_syllables = np.concatenate([v['syllable'] for v in results.values()])
valid_syllables = all_syllables[~np.isnan(all_syllables)]
valid_syllables = valid_syllables[valid_syllables >= 0].astype(int)

syllable_counts = Counter(valid_syllables)
top_syllables = [s for s, _ in syllable_counts.most_common(20)]

print("\nTop 20 most common syllables (RI1 + RI2 combined):")
for s in top_syllables:
    print(f"  Syllable {s}: {syllable_counts[s]} frames")

# STEP 13 (optional): Plot trajectory examples for top syllables
kpms.generate_trajectory_plots(
    coordinates,
    results,
    project_dir,
    model_name,
    syllables=top_syllables,
    **config()
)
```

```python
import os
import pandas as pd
import numpy as np
from tqdm import tqdm

def save_results_as_csv_with_all_keypoints(results, coordinates, project_dir, model_name, bodyparts):
    output_dir = os.path.join(project_dir, model_name, 'results_all_keypoints')
    os.makedirs(output_dir, exist_ok=True)

    print(f"Saving full keypoint CSVs to {output_dir}")

    for recording_name, result in tqdm(results.items()):
        if recording_name not in coordinates:
            print(f"Skipping {recording_name} (not found in coordinates)")
            continue
        
        n_frames = len(result['syllable'])
        coord_array = coordinates[recording_name][:n_frames]

        rows = []
        for frame in range(n_frames):
            row = {
                "recording": recording_name,
                "frame": frame,
                "syllable": result['syllable'][frame]
            }
            for b_idx, bodypart in enumerate(bodyparts):
                row[f"{bodypart}_x"] = coord_array[frame, b_idx, 0]
                row[f"{bodypart}_y"] = coord_array[frame, b_idx, 1]
            rows.append(row)

        df = pd.DataFrame(rows)
        out_file = os.path.join(output_dir, f"{recording_name}.csv")
        df.to_csv(out_file, index=False)

    print(f"✔ All CSVs saved under: {output_dir}")
```

```python
import pandas as pd
import numpy as np
import keypoint_moseq as kpms
import os
import glob
from collections import Counter

project_dir = r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2"
config = lambda: kpms.load_config(project_dir)

excel_output_dir = os.path.join(project_dir, "Excel_Results")
if not os.path.exists(excel_output_dir):
    os.makedirs(excel_output_dir)
    print(f"Created folder: {excel_output_dir}")

model_name = "2025_08_04-15_38_34"
print(f"Using model: {model_name}")

model, data, metadata, current_iter = kpms.load_checkpoint(project_dir, model_name)
print(f"Loaded checkpoint at iteration: {current_iter}")

try:
    results_path = os.path.join(project_dir, model_name, "results.h5")
    if os.path.exists(results_path):
        print("Loading existing results...")
        results = kpms.load_results(project_dir, model_name)
    else:
        print("No existing results found, extracting new results...")
        results = kpms.extract_results(model, metadata, project_dir, model_name)
except:
    print("Loading failed, extracting results with overwrite...")
    results_path = os.path.join(project_dir, model_name, "results.h5")
    if os.path.exists(results_path):
        os.remove(results_path)
    results = kpms.extract_results(model, metadata, project_dir, model_name)

print(f"Successfully loaded results for {len(results)} recordings")

def get_top_syllables_overall(results, top_n=20):
    all_syllables = []
    
    for video_key, result_data in results.items():
        syllables = result_data['syllable']
        valid_mask = ~np.isnan(syllables)
        
        if np.any(valid_mask):
            clean_syllables = syllables[valid_mask].astype(int)
            all_syllables.extend(clean_syllables)
    
    syllable_counts = Counter(all_syllables)
    top_syllables = [s for s, _ in syllable_counts.most_common(top_n)]
    
    return top_syllables

def calculate_per_video_long_bout_stats(results, top_syllables, fps=30):
    per_video_stats = []
    overall_syllable_data = {}
    
    for video_key, result_data in results.items():
        if 'track0' in video_key.lower():
            track_num = 0
        elif 'track1' in video_key.lower():
            track_num = 1
        else:
            track_num = 0
            
        video_name = video_key.split('\\')[-1].split('.')[0] if '\\' in video_key else video_key.split('.')[0]
        video_name_clean = video_name.replace('_track0', '').replace('_track1', '').replace('track0', '').replace('track1', '')
        
        syllables = result_data['syllable']
        valid_mask = ~np.isnan(syllables)
        
        if np.any(valid_mask):
            clean_syllables = syllables[valid_mask].astype(int)
            
            syllable_bout_data = {}
            durations = []
            syllable_ids = []
            current_syl = clean_syllables[0]
            start_idx = 0
            
            for i in range(1, len(clean_syllables)):
                if clean_syllables[i] != current_syl:
                    duration = i - start_idx
                    durations.append(duration)
                    syllable_ids.append(current_syl)
                    current_syl = clean_syllables[i]
                    start_idx = i
            
            final_dur = len(clean_syllables) - start_idx
            durations.append(final_dur)
            syllable_ids.append(current_syl)
            
            for syl_id, dur in zip(syllable_ids, durations):
                if syl_id not in syllable_bout_data:
                    syllable_bout_data[syl_id] = []
                syllable_bout_data[syl_id].append(dur)
                
                if syl_id not in overall_syllable_data:
                    overall_syllable_data[syl_id] = []
                overall_syllable_data[syl_id].append(dur)
            
            for syl_id in top_syllables:
                if syl_id in syllable_bout_data:
                    bout_durations = syllable_bout_data[syl_id]
                    total_bouts = len(bout_durations)
                    long_bouts = sum(1 for dur in bout_durations if dur > fps)
                    percent_long = (long_bouts / total_bouts) * 100 if total_bouts > 0 else 0
                    
                    per_video_stats.append({
                        'video_name': video_name_clean,
                        'track_number': track_num,
                        'syllable_id': syl_id,
                        'total_bouts': total_bouts,
                        'bouts_longer_than_1sec': long_bouts,
                        'percent_longer_than_1sec': percent_long,
                        'avg_duration_frames': np.mean(bout_durations),
                        'avg_duration_seconds': np.mean(bout_durations) / fps
                    })
    
    return per_video_stats, overall_syllable_data

def calculate_overall_stats(overall_syllable_data, top_syllables, fps=30):
    overall_stats = []
    
    for syl_id in top_syllables:
        if syl_id in overall_syllable_data:
            all_durations = overall_syllable_data[syl_id]
            total_bouts = len(all_durations)
            long_bouts = sum(1 for dur in all_durations if dur > fps)
            percent_long = (long_bouts / total_bouts) * 100 if total_bouts > 0 else 0
            
            overall_stats.append({
                'syllable_id': syl_id,
                'total_bouts': total_bouts,
                'bouts_longer_than_1sec': long_bouts,
                'percent_longer_than_1sec': percent_long,
                'avg_duration_frames': np.mean(all_durations),
                'avg_duration_seconds': np.mean(all_durations) / fps
            })
    
    return overall_stats

print("Finding top syllables overall...")
top_syllables = get_top_syllables_overall(results, top_n=20)
print(f"Top 20 syllables: {top_syllables}")

print("Calculating per-video and overall long bout statistics...")
per_video_stats, overall_syllable_data = calculate_per_video_long_bout_stats(results, top_syllables, fps=30)
overall_stats = calculate_overall_stats(overall_syllable_data, top_syllables, fps=30)

per_video_df = pd.DataFrame(per_video_stats)
overall_df = pd.DataFrame(overall_stats)

per_video_path = os.path.join(excel_output_dir, f"{model_name}_per_video_top_syllable_long_bout_percentages.xlsx")
overall_path = os.path.join(excel_output_dir, f"{model_name}_overall_top_syllable_long_bout_percentages.xlsx")

per_video_df.to_excel(per_video_path, index=False)
overall_df.to_excel(overall_path, index=False)

print(f"\nSaved per-video stats: Excel_Results/{model_name}_per_video_top_syllable_long_bout_percentages.xlsx")
print(f"Saved overall stats: Excel_Results/{model_name}_overall_top_syllable_long_bout_percentages.xlsx")

print("\n" + "="*80)
print("OVERALL TOP SYLLABLE LONG BOUT PERCENTAGES (>1 SECOND)")
print("="*80)

for _, row in overall_df.iterrows():
    print(f"Syllable {int(row['syllable_id'])}: {row['total_bouts']} total bouts, {row['percent_longer_than_1sec']:.1f}% are >1sec")
    print(f"  ({row['bouts_longer_than_1sec']} out of {row['total_bouts']} bouts, avg duration: {row['avg_duration_seconds']:.2f} sec)")

print("\n" + "="*80)
print("PER-VIDEO TOP SYLLABLE LONG BOUT PERCENTAGES (>1 SECOND)")
print("="*80)

for video_name in per_video_df['video_name'].unique():
    print(f"\n{video_name}:")
    video_data = per_video_df[per_video_df['video_name'] == video_name]
    
    for track_num in sorted(video_data['track_number'].unique()):
        print(f"  Track {track_num}:")
        track_data = video_data[video_data['track_number'] == track_num]
        
        for _, row in track_data.iterrows():
            print(f"    Syllable {int(row['syllable_id'])}: {row['percent_longer_than_1sec']:.1f}% of {row['total_bouts']} bouts >1sec")

print(f"\nFiles saved to: {excel_output_dir}")
```

```python
import os
import h5py
import numpy as np
from scipy import signal
import pandas as pd
import matplotlib.pyplot as plt

def process_respiratory_file(file_path, original_fs=20000, target_fs=100):
    """
    Process a single respiratory h5 file: filter and calculate breathing rate
    """
    print(f"Processing: {os.path.basename(file_path)}")
    
    # Load respiratory data
    with h5py.File(file_path, 'r') as f:
        resp_data = f['resp'][:].flatten()  # Remove extra dimension
    
    print(f"  Original data: {len(resp_data)} samples at {original_fs}Hz")
    print(f"  Duration: {len(resp_data)/original_fs:.2f} seconds")
    
    # Design low-pass filter for 100Hz cutoff
    nyquist = original_fs / 2
    cutoff = target_fs / 2  
    b, a = signal.butter(5, cutoff/nyquist, 'low')
    
    # Filter the data
    filtered_resp = signal.filtfilt(b, a, resp_data)
    
    # Downsample to 100Hz
    downsample_factor = original_fs // target_fs  # 200
    downsampled_resp = filtered_resp[::downsample_factor]
    
    print(f"  Filtered data: {len(downsampled_resp)} samples at {target_fs}Hz")
    
    # Calculate breathing rate (1/IBI)
    breathing_rate, breath_times = calculate_breathing_rate(downsampled_resp, target_fs)
    
    return downsampled_resp, breathing_rate, breath_times

def calculate_breathing_rate(resp_signal, fs=100):
    """
    Calculate 1/IBI (breathing rate) from respiratory signal
    """
    # Normalize signal for peak detection
    normalized_resp = (resp_signal - np.mean(resp_signal)) / np.std(resp_signal)
    
    # Find breathing peaks (inspiration peaks)
    # Adjust parameters based on your breathing rate expectations
    min_distance = fs // 4  # Minimum 0.25s between breaths (240 BPM max)
    prominence = 0.5  # Adjust based on signal quality
    
    peaks, properties = signal.find_peaks(normalized_resp, 
                                        distance=min_distance,
                                        prominence=prominence,
                                        height=0.1)
    
    print(f"  Found {len(peaks)} breathing peaks")
    
    if len(peaks) < 2:
        print("  Warning: Not enough peaks found for IBI calculation")
        return np.array([]), np.array([])
    
    # Calculate inter-breath intervals (IBI)
    ibi_samples = np.diff(peaks)  # IBI in samples
    ibi_seconds = ibi_samples / fs  # IBI in seconds
    breathing_rate = 1 / ibi_seconds  # 1/IBI (breaths per second)
    
    # Convert to breaths per minute for interpretation
    breaths_per_minute = breathing_rate * 60
    
    # Timestamps for breathing rate (at each breath after the first)
    breath_times = peaks[1:] / fs
    
    print(f"  Mean breathing rate: {np.mean(breaths_per_minute):.1f} breaths/min")
    print(f"  Range: {np.min(breaths_per_minute):.1f} - {np.max(breaths_per_minute):.1f} breaths/min")
    
    return breathing_rate, breath_times

def process_all_respiratory_files(resp_dir):
    """
    Process all respiratory h5 files in directory
    """
    h5_files = [f for f in os.listdir(resp_dir) if f.endswith('.h5')]
    
    results = {}
    
    for h5_file in h5_files:
        file_path = os.path.join(resp_dir, h5_file)
        
        try:
            filtered_resp, breathing_rate, breath_times = process_respiratory_file(file_path)
            
            results[h5_file] = {
                'filtered_signal': filtered_resp,
                'breathing_rate': breathing_rate,
                'breath_times': breath_times,
                'file_path': file_path
            }
        except Exception as e:
            print(f"Error processing {h5_file}: {e}")
    
    return results

# Main processing
resp_path = r"C:\Users\Padilla-Coreano\Documents\tanish\Resp_h5"
respiratory_data = process_all_respiratory_files(resp_path)
```

```python
import pandas as pd
import numpy as np
import glob
import os
from scipy import signal
import h5py

def process_syllable_instances(resp_data, excel_file_path):
    """Process respiratory data per syllable instance, not per frame"""
    
    # Load syllable data
    syllable_df = pd.read_excel(excel_file_path)
    print(f"  Loaded syllable data: {syllable_df.shape}")
    
    # Get FPS and convert frames to time
    video_fps = get_fps_for_file(os.path.basename(excel_file_path))
    syllable_df['time_seconds'] = syllable_df['frame_number'] / video_fps
    
    print(f"  Using {video_fps} FPS for time conversion")
    
    # Group consecutive frames with same syllable to find syllable instances
    syllable_df['syllable_group'] = (syllable_df['syllable'] != syllable_df['syllable'].shift()).cumsum()
    
    syllable_instances = []
    
    for group_id, group in syllable_df.groupby('syllable_group'):
        syllable_id = group['syllable'].iloc[0]
        start_frame = group['frame_number'].min()
        end_frame = group['frame_number'].max()
        start_time = group['time_seconds'].min()
        end_time = group['time_seconds'].max()
        duration_ms = (end_time - start_time) * 1000  # Convert to milliseconds
        
        # Skip syllables under 500ms
        if duration_ms < 500:
            continue
            
        num_frames = len(group)
        
        # Get respiratory data for this syllable duration
        # Find respiratory data points within this time window
        time_mask = (resp_data['time_vector'] >= start_time) & (resp_data['time_vector'] <= end_time)
        syllable_resp_signal = resp_data['filtered_signal'][time_mask]
        
        if len(syllable_resp_signal) == 0:
            continue
            
        # Calculate respiratory metrics for this syllable instance
        mean_resp_signal = np.mean(syllable_resp_signal)
        std_resp_signal = np.std(syllable_resp_signal)
        min_resp_signal = np.min(syllable_resp_signal)
        max_resp_signal = np.max(syllable_resp_signal)
        
        # Get breathing rate during this syllable
        # Find breath times that fall within this syllable
        if len(resp_data['breathing_rate']) > 0:
            breath_mask = (resp_data['breath_times'] >= start_time) & (resp_data['breath_times'] <= end_time)
            syllable_breathing_rates = resp_data['breathing_rate'][breath_mask]
            
            if len(syllable_breathing_rates) > 0:
                mean_breathing_rate_per_sec = np.mean(syllable_breathing_rates)  # Already in breaths/sec
                mean_breathing_rate_bpm = mean_breathing_rate_per_sec * 60
                std_breathing_rate_per_sec = np.std(syllable_breathing_rates)
                
                # Calculate IBI for this syllable
                mean_ibi = 1 / mean_breathing_rate_per_sec if mean_breathing_rate_per_sec > 0 else np.nan
            else:
                # No breathing events during this syllable, interpolate from nearest
                interp_rate = np.interp((start_time + end_time) / 2, 
                                      resp_data['breath_times'], 
                                      resp_data['breathing_rate'])
                mean_breathing_rate_per_sec = interp_rate
                mean_breathing_rate_bpm = interp_rate * 60
                std_breathing_rate_per_sec = np.nan
                mean_ibi = 1 / interp_rate if interp_rate > 0 else np.nan
        else:
            mean_breathing_rate_per_sec = np.nan
            mean_breathing_rate_bpm = np.nan
            std_breathing_rate_per_sec = np.nan
            mean_ibi = np.nan
        
        syllable_instances.append({
            'syllable_id': syllable_id,
            'start_frame': start_frame,
            'end_frame': end_frame,
            'start_time_sec': start_time,
            'end_time_sec': end_time,
            'duration_ms': duration_ms,
            'num_frames': num_frames,
            'fps_used': video_fps,
            'mean_resp_signal_100hz': mean_resp_signal,
            'std_resp_signal_100hz': std_resp_signal,
            'min_resp_signal_100hz': min_resp_signal,
            'max_resp_signal_100hz': max_resp_signal,
            'mean_breathing_rate_per_sec': mean_breathing_rate_per_sec,
            'mean_breathing_rate_bpm': mean_breathing_rate_bpm,
            'std_breathing_rate_per_sec': std_breathing_rate_per_sec,
            'mean_ibi_seconds': mean_ibi,
            'video_name': group['video_name'].iloc[0],
            'track_name': group['track_name'].iloc[0]
        })
    
    result_df = pd.DataFrame(syllable_instances)
    print(f"  Created {len(result_df)} syllable instances (≥500ms) from {len(syllable_df)} frames")
    
    if len(result_df) > 0:
        print(f"  Duration range: {result_df['duration_ms'].min():.0f} - {result_df['duration_ms'].max():.0f} ms")
        print(f"  Mean syllable duration: {result_df['duration_ms'].mean():.0f} ms")
    
    return result_df

def process_track0_only_with_syllable_instances(resp_dir, excel_dir):
    """Process only track0 files with syllable instance analysis"""
    
    h5_files = glob.glob(os.path.join(resp_dir, "*.h5"))
    excel_files = glob.glob(os.path.join(excel_dir, "*.xlsx")) + glob.glob(os.path.join(excel_dir, "*.xls"))
    
    # Filter to only track0 files
    track0_excel_files = [f for f in excel_files if "track0" in f]
    
    print(f"Found {len(h5_files)} h5 files and {len(track0_excel_files)} track0 Excel files")
    
    results = {}
    
    for h5_file in h5_files:
        h5_basename = os.path.basename(h5_file).replace('.h5', '')
        print(f"\n=== Processing {h5_basename} ===")
        
        # Process respiratory data
        resp_data = process_respiratory_file(h5_file)
        
        # Find matching track0 Excel files only
        matching_excel_files = [f for f in track0_excel_files 
                               if match_files_by_name(h5_basename, [f])]
        
        if matching_excel_files:
            for excel_file in matching_excel_files:
                print(f"  Processing: {os.path.basename(excel_file)}")
                syllable_instances_df = process_syllable_instances(resp_data, excel_file)
                
                if syllable_instances_df is not None and len(syllable_instances_df) > 0:
                    # Create output filename
                    excel_basename = os.path.basename(excel_file).replace('.xlsx', '').replace('.xls', '')
                    output_file = f"SYLLABLE_INSTANCES_{excel_basename}_with_respiration.xlsx"
                    
                    syllable_instances_df.to_excel(output_file, index=False)
                    print(f"  ✓ Saved: {output_file}")
                    
                    # Show stats
                    valid_bpm = syllable_instances_df['mean_breathing_rate_bpm'].dropna()
                    if len(valid_bpm) > 0:
                        print(f"  Breathing rate: {valid_bpm.mean():.1f} ± {valid_bpm.std():.1f} BPM")
                        print(f"  Breathing rate: {valid_bpm.mean()/60:.3f} ± {valid_bpm.std()/60:.3f} breaths/sec")
                    
                    results[f"{h5_basename}_{excel_basename}"] = syllable_instances_df
                else:
                    print(f"  No valid syllable instances found (all <500ms)")
        else:
            print(f"  No matching track0 Excel files found")
    
    return results

# Updated processing
resp_path = r"C:\Users\Padilla-Coreano\Documents\tanish\Resp_h5"
excel_path = r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results"

print("Processing TRACK0 ONLY with SYLLABLE INSTANCES (≥500ms)...")
syllable_results = process_track0_only_with_syllable_instances(resp_path, excel_path)

print(f"\n🎉 SYLLABLE INSTANCE PROCESSING COMPLETE! 🎉")
print(f"Generated {len(syllable_results)} files")

print(f"\nYour new files contain:")
print(f"  ✓ One row per syllable instance (not per frame)")
print(f"  ✓ Only syllables lasting ≥500ms")
print(f"  ✓ Only track0 data")
print(f"  ✓ Breathing rate in breaths/second AND BPM")
print(f"  ✓ IBI (Inter-Breath Interval) in seconds")
print(f"  ✓ Respiratory signal statistics per syllable")
```

```python
import h5py
import numpy as np
import pandas as pd
from scipy import signal
import glob
import os
import matplotlib.pyplot as plt

def extract_breathing_signal(h5_file_path):
    with h5py.File(h5_file_path, 'r') as f:
        breathing_data = f['resp'][:].flatten()
        print(f"Extracted breathing signal: {len(breathing_data)} samples")
        print(f"Signal range: {breathing_data.min():.1f} to {breathing_data.max():.1f}")
    return breathing_data

def apply_signal_filtering(breathing_data, source_fs=20000, output_fs=100):
    print(f"Source sampling rate: {source_fs} Hz")
    print(f"Output sampling rate: {output_fs} Hz")
    
    nyquist_freq = source_fs / 2
    filter_cutoff = output_fs / 2
    
    filter_b, filter_a = signal.butter(5, filter_cutoff/nyquist_freq, 'low')
    
    clean_signal = signal.filtfilt(filter_b, filter_a, breathing_data)
    print(f"Signal filtered with cutoff: {filter_cutoff} Hz")
    
    reduction_factor = source_fs // output_fs
    reduced_signal = clean_signal[::reduction_factor]
    
    print(f"Signal reduced by factor {reduction_factor}: {len(reduced_signal)} samples at {output_fs} Hz")
    print(f"Total duration: {len(reduced_signal) / output_fs:.2f} seconds")
    
    return reduced_signal, output_fs

def generate_timestamp_array(clean_signal, sample_rate):
    timestamp_array = np.arange(len(clean_signal)) / sample_rate
    print(f"Timestamp array: 0.00 to {timestamp_array[-1]:.2f} seconds")
    return timestamp_array

def find_breath_cycles(breathing_signal, fs=100):
    standardized_signal = (breathing_signal - np.mean(breathing_signal)) / np.std(breathing_signal)
    
    min_peak_distance = fs // 4
    peak_prominence = 0.5
    min_height = 0.1
    
    breath_peaks, peak_properties = signal.find_peaks(standardized_signal,
                                                     distance=min_peak_distance,
                                                     prominence=peak_prominence,
                                                     height=min_height)
    
    print(f"Detected {len(breath_peaks)} breath cycles")
    
    if len(breath_peaks) < 2:
        print("Warning: Insufficient breath cycles for interval calculation")
        return np.array([]), np.array([]), np.array([])
    
    cycle_timestamps = breath_peaks / fs
    interval_durations = np.diff(cycle_timestamps)
    interval_timestamps = cycle_timestamps[1:]
    
    print(f"Breath interval range: {interval_durations.min():.3f} to {interval_durations.max():.3f} seconds")
    print(f"Average breath interval: {interval_durations.mean():.3f} seconds")
    
    return breath_peaks, interval_durations, interval_timestamps

def compute_breath_frequencies(interval_durations, interval_timestamps):
    frequency_per_second = 1 / interval_durations
    
    frequency_per_minute = frequency_per_second * 60
    
    print(f"Breath frequency range: {frequency_per_second.min():.3f} to {frequency_per_second.max():.3f} breaths/sec")
    print(f"Breath frequency range: {frequency_per_minute.min():.1f} to {frequency_per_minute.max():.1f} breaths/min")
    print(f"Average breath frequency: {frequency_per_second.mean():.3f} breaths/sec ({frequency_per_minute.mean():.1f} breaths/min)")
    
    return frequency_per_second, frequency_per_minute

def determine_video_framerate(filename):
    framerate_mapping = {
        "RI1_s1_1": 15,
        "RI1_s2_3": 15, 
        "RI1_s3_6": 29,
        "RI2_s1_1": 15,
        "RI2_s2_4": 15
    }
    
    for pattern, fps in framerate_mapping.items():
        if pattern in filename:
            return fps
    return 30

def import_behavior_data_with_timing(excel_file_path):
    behavior_df = pd.read_excel(excel_file_path)
    print(f"Imported behavior data: {behavior_df.shape}")
    print(f"Data columns: {list(behavior_df.columns)}")
    
    video_framerate = determine_video_framerate(os.path.basename(excel_file_path))
    print(f"Applied {video_framerate} FPS for frame-to-time conversion")
    
    behavior_df['timestamp_seconds'] = behavior_df['frame_number'] / video_framerate
    behavior_df['framerate_applied'] = video_framerate
    
    print(f"Behavior timestamp range: {behavior_df['timestamp_seconds'].min():.2f} to {behavior_df['timestamp_seconds'].max():.2f} seconds")
    
    return behavior_df, video_framerate

def create_behavior_episodes(behavior_df, minimum_duration_ms=500):
    behavior_df['behavior_transition'] = (behavior_df['syllable'] != behavior_df['syllable'].shift()).astype(int)
    behavior_df['episode_group'] = behavior_df['behavior_transition'].cumsum()
    
    behavior_episodes = []
    
    for group_id, episode_group in behavior_df.groupby('episode_group'):
        behavior_type = episode_group['syllable'].iloc[0]
        episode_start_frame = episode_group['frame_number'].min()
        episode_end_frame = episode_group['frame_number'].max()
        episode_start_time = episode_group['timestamp_seconds'].min()
        episode_end_time = episode_group['timestamp_seconds'].max()
        episode_duration_ms = (episode_end_time - episode_start_time) * 1000
        episode_frame_count = len(episode_group)
        applied_framerate = episode_group['framerate_applied'].iloc[0]
        
        if episode_duration_ms >= minimum_duration_ms:
            behavior_episodes.append({
                'behavior_type': behavior_type,
                'episode_start_frame': episode_start_frame,
                'episode_end_frame': episode_end_frame,
                'episode_start_time': episode_start_time,
                'episode_end_time': episode_end_time,
                'episode_duration_ms': episode_duration_ms,
                'episode_frame_count': episode_frame_count,
                'framerate_applied': applied_framerate,
                'source_video': episode_group['video_name'].iloc[0] if 'video_name' in episode_group.columns else 'unknown',
                'tracking_id': episode_group['track_name'].iloc[0] if 'track_name' in episode_group.columns else 'track0'
            })
    
    episodes_df = pd.DataFrame(behavior_episodes)
    print(f"Generated {len(episodes_df)} behavior episodes ≥{minimum_duration_ms}ms from {len(behavior_df)} frames")
    
    if len(episodes_df) > 0:
        print(f"Episode duration range: {episodes_df['episode_duration_ms'].min():.0f} - {episodes_df['episode_duration_ms'].max():.0f} ms")
    
    return episodes_df

def synchronize_breathing_with_behaviors(episodes_df, breathing_timestamps, breathing_signal, 
                                       breath_frequency_per_sec, breath_frequency_per_min, 
                                       breath_intervals, interval_timestamps):
    synchronized_data = []
    
    for idx, episode in episodes_df.iterrows():
        episode_start = episode['episode_start_time']
        episode_end = episode['episode_end_time']
        
        time_window = (breathing_timestamps >= episode_start) & (breathing_timestamps <= episode_end)
        episode_breathing_signal = breathing_signal[time_window]
        
        if len(episode_breathing_signal) > 0:
            signal_mean = np.mean(episode_breathing_signal)
            signal_std = np.std(episode_breathing_signal)
            signal_min = np.min(episode_breathing_signal)
            signal_max = np.max(episode_breathing_signal)
            signal_span = signal_max - signal_min
        else:
            signal_mean = signal_std = signal_min = signal_max = signal_span = np.nan
        
        breath_window = (interval_timestamps >= episode_start) & (interval_timestamps <= episode_end)
        episode_frequencies_per_sec = breath_frequency_per_sec[breath_window]
        episode_frequencies_per_min = breath_frequency_per_min[breath_window]
        episode_intervals = breath_intervals[breath_window]
        
        if len(episode_frequencies_per_sec) > 0:
            avg_frequency_per_sec = np.mean(episode_frequencies_per_sec)
            avg_frequency_per_min = np.mean(episode_frequencies_per_min)
            frequency_std_per_sec = np.std(episode_frequencies_per_sec)
            avg_breath_interval = np.mean(episode_intervals)
            interval_std = np.std(episode_intervals)
            breath_count = len(episode_frequencies_per_sec)
        else:
            episode_midpoint = (episode_start + episode_end) / 2
            if len(interval_timestamps) > 0:
                avg_frequency_per_sec = np.interp(episode_midpoint, interval_timestamps, breath_frequency_per_sec)
                avg_frequency_per_min = avg_frequency_per_sec * 60
                avg_breath_interval = 1 / avg_frequency_per_sec if avg_frequency_per_sec > 0 else np.nan
            else:
                avg_frequency_per_sec = avg_frequency_per_min = avg_breath_interval = np.nan
            
            frequency_std_per_sec = interval_std = np.nan
            breath_count = 0
        
        episode_data = {
            **episode.to_dict(),
            'avg_breathing_signal_100hz': signal_mean,
            'std_breathing_signal_100hz': signal_std,
            'min_breathing_signal_100hz': signal_min,
            'max_breathing_signal_100hz': signal_max,
            'breathing_signal_span': signal_span,
            'avg_breath_frequency_per_sec': avg_frequency_per_sec,
            'avg_breath_frequency_per_min': avg_frequency_per_min,
            'std_breath_frequency_per_sec': frequency_std_per_sec,
            'avg_breath_interval_seconds': avg_breath_interval,
            'std_breath_interval_seconds': interval_std,
            'complete_breath_count': breath_count
        }
        
        synchronized_data.append(episode_data)
    
    synchronized_df = pd.DataFrame(synchronized_data)
    print(f"Synchronized breathing data with {len(synchronized_df)} behavior episodes")
    
    return synchronized_df

def execute_full_synchronization_pipeline(h5_file, excel_file, minimum_duration_ms=500):
    print(f"=== Executing synchronization for {os.path.basename(h5_file)} with {os.path.basename(excel_file)} ===")
    
    raw_breathing_data = extract_breathing_signal(h5_file)
    
    filtered_breathing, target_sampling_rate = apply_signal_filtering(raw_breathing_data)
    
    breathing_timestamps = generate_timestamp_array(filtered_breathing, target_sampling_rate)
    
    breath_peaks, breath_intervals, interval_timestamps = find_breath_cycles(filtered_breathing, target_sampling_rate)
    
    if len(breath_intervals) > 0:
        breath_frequency_per_sec, breath_frequency_per_min = compute_breath_frequencies(breath_intervals, interval_timestamps)
    else:
        breath_frequency_per_sec = breath_frequency_per_min = np.array([])
    
    behavior_df, video_framerate = import_behavior_data_with_timing(excel_file)
    
    episodes_df = create_behavior_episodes(behavior_df, minimum_durationh5_ms)
    
    if len(episodes_df) > 0 and len(breath_intervals) > 0:
        final_synchronized_df = synchronize_breathing_with_behaviors(episodes_df, breathing_timestamps, filtered_breathing,
                                                   breath_frequency_per_sec, breath_frequency_per_min,
                                                   breath_intervals, interval_timestamps)
    else:
        print("No valid data for synchronization")
        return None
    
    h5_base_name = os.path.basename(h5_file).replace('.h5', '')
    excel_base_name = os.path.basename(excel_file).replace('.xlsx', '').replace('.xls', '')
    output_filename = f"SYNCHRONIZED_{excel_base_name}_breathing_data.xlsx"
    
    final_synchronized_df.to_excel(output_filename, index=False)
    print(f"✅ Saved synchronized data: {output_filename}")
    
    if len(final_synchronized_df) > 0:
        valid_frequencies = final_synchronized_df['avg_breath_frequency_per_min'].dropna()
        if len(valid_frequencies) > 0:
            print(f"Summary - Average breath frequency: {valid_frequencies.mean():.1f} breaths/min")
            print(f"Summary - Breath frequency range: {valid_frequencies.min():.1f} - {valid_frequencies.max():.1f} breaths/min")
    
    return final_synchronized_df

def execute_complete_processing_pipeline():
    breathing_data_path = r"C:\Users\Padilla-Coreano\Documents\tanish\Resp_h5"
    behavior_data_path = r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results"
    
    h5_file_list = glob.glob(os.path.join(breathing_data_path, "*.h5"))
    excel_track0_files = glob.glob(os.path.join(behavior_data_path, "*track0*.xlsx"))
    
    print(f"Located {len(h5_file_list)} h5 files and {len(excel_track0_files)} track0 Excel files")
    
    processing_results = {}
    
    for h5_file in h5_file_list:
        h5_identifier = os.path.basename(h5_file).replace('.h5', '')
        
        matched_excel = None
        for excel_file in excel_track0_files:
            identifier_parts = h5_identifier.split('_')
            if len(identifier_parts) >= 3:
                session_identifier = '_'.join(identifier_parts[:3])
                if session_identifier in excel_file:
                    matched_excel = excel_file
                    break
        
        if matched_excel:
            try:
                processing_result = execute_full_synchronization_pipeline(h5_file, matched_excel)
                if processing_result is not None:
                    processing_results[h5_identifier] = processing_result
            except Exception as e:
                print(f"Error processing {h5_identifier}: {e}")
        else:
            print(f"No matching Excel file found for {h5_identifier}")
    
    print(f"\n🎉 COMPLETE PROCESSING PIPELINE FINISHED! 🎉")
    print(f"Successfully processed {len(processing_results)} file pairs")
    
    return processing_results

all_processing_results = execute_complete_processing_pipeline()
```

```python
import h5py
import numpy as np
import pandas as pd
from scipy import signal
import glob
import os
import matplotlib.pyplot as plt

def load_respiratory_data(h5_file_path):
    with h5py.File(h5_file_path, 'r') as f:
        data = f['resp'][:].flatten()
        print(f"Loaded respiratory data: {len(data)} samples")
        print(f"Data range: {data.min():.1f} to {data.max():.1f}")
    return data

def filter_respiratory_signal(data, original_fs=20000, target_fs=100):
    print(f"Original sampling rate: {original_fs} Hz")
    print(f"Target sampling rate: {target_fs} Hz")
    
    nyquist = original_fs / 2
    cutoff = target_fs / 2
    
    b, a = signal.butter(5, cutoff/nyquist, 'low')
    filtered = signal.filtfilt(b, a, data)
    print(f"Applied filter with cutoff: {cutoff} Hz")
    
    downsample_factor = original_fs // target_fs
    downsampled = filtered[::downsample_factor]
    
    print(f"Downsampled by factor {downsample_factor}: {len(downsampled)} samples at {target_fs} Hz")
    print(f"Duration: {len(downsampled) / target_fs:.2f} seconds")
    
    return downsampled, target_fs

def create_time_vector(signal_data, fs):
    time_vec = np.arange(len(signal_data)) / fs
    print(f"Time vector: 0.00 to {time_vec[-1]:.2f} seconds")
    return time_vec

def detect_breathing_events(signal_data, fs=100):
    normalized = (signal_data - np.mean(signal_data)) / np.std(signal_data)
    
    min_distance = fs // 4
    prominence = 0.5
    height = 0.1
    
    peaks, _ = signal.find_peaks(normalized, distance=min_distance, prominence=prominence, height=height)
    
    print(f"Found {len(peaks)} breathing events")
    
    if len(peaks) < 2:
        print("Warning: Not enough events detected")
        return np.array([]), np.array([]), np.array([])
    
    peak_times = peaks / fs
    intervals = np.diff(peak_times)
    interval_times = peak_times[1:]
    
    print(f"Interval range: {intervals.min():.3f} to {intervals.max():.3f} seconds")
    print(f"Mean interval: {intervals.mean():.3f} seconds")
    
    return peaks, intervals, interval_times

def calculate_breathing_rates(intervals, times):
    rates_per_sec = 1 / intervals
    rates_per_min = rates_per_sec * 60
    
    print(f"Rate range: {rates_per_sec.min():.3f} to {rates_per_sec.max():.3f} breaths/sec")
    print(f"Rate range: {rates_per_min.min():.1f} to {rates_per_min.max():.1f} breaths/min")
    print(f"Mean rate: {rates_per_sec.mean():.3f} breaths/sec ({rates_per_min.mean():.1f} breaths/min)")
    
    return rates_per_sec, rates_per_min

def get_video_fps(filename):
    fps_dict = {
        "RI1_s1_1": 15,
        "RI1_s2_3": 15,
        "RI1_s3_6": 29,
        "RI2_s1_1": 15,
        "RI2_s2_4": 15
    }
    
    for key, fps in fps_dict.items():
        if key in filename:
            return fps
    return 30

def load_behavior_data(excel_path):
    df = pd.read_excel(excel_path)
    print(f"Loaded behavior data: {df.shape}")
    print(f"Columns: {list(df.columns)}")
    
    fps = get_video_fps(os.path.basename(excel_path))
    print(f"Using {fps} FPS for conversion")
    
    df['time_sec'] = df['frame_number'] / fps
    df['fps'] = fps
    
    print(f"Time range: {df['time_sec'].min():.2f} to {df['time_sec'].max():.2f} seconds")
    
    return df, fps

def group_behavior_bouts(df, min_duration=500):
    df['transition'] = (df['syllable'] != df['syllable'].shift()).astype(int)
    df['group'] = df['transition'].cumsum()
    
    bouts = []
    
    for group_id, group in df.groupby('group'):
        syllable = group['syllable'].iloc[0]
        start_frame = group['frame_number'].min()
        end_frame = group['frame_number'].max()
        start_time = group['time_sec'].min()
        end_time = group['time_sec'].max()
        duration = (end_time - start_time) * 1000
        frame_count = len(group)
        fps = group['fps'].iloc[0]
        
        if duration >= min_duration:
            bouts.append({
                'syllable': syllable,
                'start_frame': start_frame,
                'end_frame': end_frame,
                'start_time': start_time,
                'end_time': end_time,
                'duration_ms': duration,
                'frame_count': frame_count,
                'fps': fps,
                'video': group['video_name'].iloc[0] if 'video_name' in group.columns else 'unknown',
                'track': group['track_name'].iloc[0] if 'track_name' in group.columns else 'track0'
            })
    
    bout_df = pd.DataFrame(bouts)
    print(f"Created {len(bout_df)} bouts ≥{min_duration}ms from {len(df)} frames")
    
    if len(bout_df) > 0:
        print(f"Duration range: {bout_df['duration_ms'].min():.0f} - {bout_df['duration_ms'].max():.0f} ms")
    
    return bout_df

def align_respiratory_behavioral_data(bouts, time_vec, signal_data, rates_per_sec, rates_per_min, intervals, interval_times):
    results = []
    
    for idx, bout in bouts.iterrows():
        start = bout['start_time']
        end = bout['end_time']
        
        signal_mask = (time_vec >= start) & (time_vec <= end)
        bout_signal = signal_data[signal_mask]
        
        if len(bout_signal) > 0:
            sig_mean = np.mean(bout_signal)
            sig_min = np.min(bout_signal)
            sig_max = np.max(bout_signal)
            sig_range = sig_max - sig_min
        else:
            sig_mean = sig_min = sig_max = sig_range = np.nan
        
        rate_mask = (interval_times >= start) & (interval_times <= end)
        bout_rates_sec = rates_per_sec[rate_mask]
        bout_rates_min = rates_per_min[rate_mask]
        bout_intervals = intervals[rate_mask]
        
        if len(bout_rates_sec) > 0:
            mean_rate_sec = np.mean(bout_rates_sec)
            mean_rate_min = np.mean(bout_rates_min)
            mean_interval = np.mean(bout_intervals)
            breath_count = len(bout_rates_sec)
        else:
            mid_time = (start + end) / 2
            if len(interval_times) > 0:
                mean_rate_sec = np.interp(mid_time, interval_times, rates_per_sec)
                mean_rate_min = mean_rate_sec * 60
                mean_interval = 1 / mean_rate_sec if mean_rate_sec > 0 else np.nan
            else:
                mean_rate_sec = mean_rate_min = mean_interval = np.nan
            
            breath_count = 0
        
        result = {
            **bout.to_dict(),
            'mean_resp_100hz': sig_mean,
            'min_resp_100hz': sig_min,
            'max_resp_100hz': sig_max,
            'resp_range': sig_range,
            'mean_rate_per_sec': mean_rate_sec,
            'mean_rate_per_min': mean_rate_min,
            'mean_interval_sec': mean_interval,
            'breath_count': breath_count
        }
        
        results.append(result)
    
    result_df = pd.DataFrame(results)
    print(f"Aligned data for {len(result_df)} behavioral bouts")
    
    return result_df

def process_session(h5_file, excel_file, min_duration=500):
    print(f"=== Processing {os.path.basename(h5_file)} with {os.path.basename(excel_file)} ===")
    
    resp_data = load_respiratory_data(h5_file)
    filtered_resp, fs = filter_respiratory_signal(resp_data)
    time_vec = create_time_vector(filtered_resp, fs)
    peaks, intervals, interval_times = detect_breathing_events(filtered_resp, fs)
    
    if len(intervals) > 0:
        rates_sec, rates_min = calculate_breathing_rates(intervals, interval_times)
    else:
        rates_sec = rates_min = np.array([])
    
    behavior_df, fps = load_behavior_data(excel_file)
    bouts = group_behavior_bouts(behavior_df, min_duration)
    
    if len(bouts) > 0 and len(intervals) > 0:
        final_df = align_respiratory_behavioral_data(bouts, time_vec, filtered_resp, rates_sec, rates_min, intervals, interval_times)
    else:
        print("No valid data for alignment")
        return None
    
    h5_name = os.path.basename(h5_file).replace('.h5', '')
    excel_name = os.path.basename(excel_file).replace('.xlsx', '').replace('.xls', '')
    output_file = f"FINAL_PROCESSED_{excel_name}_no_std.xlsx"
    
    final_df.to_excel(output_file, index=False)
    print(f"Saved data: {output_file}")
    
    if len(final_df) > 0:
        valid_rates = final_df['mean_rate_per_min'].dropna()
        if len(valid_rates) > 0:
            print(f"Mean breathing rate: {valid_rates.mean():.1f} breaths/min")
            print(f"Rate range: {valid_rates.min():.1f} - {valid_rates.max():.1f} breaths/min")
    
    return final_df

def run_analysis():
    resp_dir = r"C:\Users\Padilla-Coreano\Documents\tanish\Resp_h5"
    behavior_dir = r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results"
    
    h5_files = glob.glob(os.path.join(resp_dir, "*.h5"))
    excel_files = glob.glob(os.path.join(behavior_dir, "*track0*.xlsx"))
    
    print(f"Found {len(h5_files)} h5 files and {len(excel_files)} Excel files")
    
    results = {}
    
    for h5_file in h5_files:
        h5_id = os.path.basename(h5_file).replace('.h5', '')
        
        matched_excel = None
        for excel_file in excel_files:
            parts = h5_id.split('_')
            if len(parts) >= 3:
                session_id = '_'.join(parts[:3])
                if session_id in excel_file:
                    matched_excel = excel_file
                    break
        
        if matched_excel:
            try:
                result = process_session(h5_file, matched_excel)
                if result is not None:
                    results[h5_id] = result
            except Exception as e:
                print(f"Error processing {h5_id}: {e}")
        else:
            print(f"No matching file found for {h5_id}")
    
    print(f"\nAnalysis complete")
    print(f"Processed {len(results)} sessions")
    
    return results

def create_master_files_no_std():
    processed_files = glob.glob("FINAL_PROCESSED_*_no_std.xlsx")
    
    ri1_files = [f for f in processed_files if "-RI1_" in f]
    ri2_files = [f for f in processed_files if "-RI2_" in f]
    
    print(f"Found {len(ri1_files)} RI1 files and {len(ri2_files)} RI2 files")
    
    def create_master_file(files, output_name):
        print(f"\nCreating {output_name}...")
        
        with pd.ExcelWriter(output_name, engine='openpyxl') as writer:
            for file in files:
                df = pd.read_excel(file)
                
                filename = os.path.basename(file)
                
                if "-RI1_" in filename:
                    session_part = filename.split("-RI1_")[1]
                    sheet_prefix = "RI1_"
                elif "-RI2_" in filename:
                    session_part = filename.split("-RI2_")[1]
                    sheet_prefix = "RI2_"
                
                session_info = session_part.split("_")[0:3]
                
                sheet_name = f"{sheet_prefix}{'_'.join(session_info)}_track0"
                
                if len(sheet_name) > 31:
                    sheet_name = sheet_name[:31]
                
                df.to_excel(writer, sheet_name=sheet_name, index=False)
                print(f"  Added sheet: {sheet_name} ({df.shape[0]} behavioral bouts)")
        
        print(f"Saved: {output_name}")
    
    if ri1_files:
        create_master_file(ri1_files, "RI1_MASTER_NoStd_Respiratory_Behavioral.xlsx")
    
    if ri2_files:
        create_master_file(ri2_files, "RI2_MASTER_NoStd_Respiratory_Behavioral.xlsx")
    
    return len(ri1_files), len(ri2_files)

analysis_results = run_analysis()

ri1_count, ri2_count = create_master_files_no_std()

print(f"\nMaster files created without standard deviations")
print(f"RI1_MASTER_NoStd_Respiratory_Behavioral.xlsx ({ri1_count} sheets)")
print(f"RI2_MASTER_NoStd_Respiratory_Behavioral.xlsx ({ri2_count} sheets)")
```

```python
import h5py
import numpy as np
from scipy.signal import butter, filtfilt, resample
import neurokit2 as nk
import os

def load_clean_resp_signal(h5_file, target_rate=100):

    try:
        with h5py.File(h5_file, 'r') as f:
            resp = f['resp'][:].flatten()
            ekg_meta = dict(f['ekg_metadata'].attrs)
        duration_sec = float(ekg_meta['duration_sec'])
        fs = len(resp) / duration_sec
        print(f"Loaded {os.path.basename(h5_file)}: {len(resp)} samples, {fs:.2f} Hz, {duration_sec:.2f} sec")
    except Exception as e:
        print(f"Error loading {h5_file}: {e}")
        return None, None, None, None
    
    # Process signal
    filtered = apply_lowpass_filter(resp, fs, target_rate)
    resampled = resample_signal(filtered, duration_sec, target_rate)
    cleaned = apply_bandpass_filter(resampled, target_rate)
    time_vector = np.arange(len(cleaned)) / float(target_rate)
    
    return cleaned, time_vector, float(target_rate), os.path.basename(h5_file)
```

```python
#Low pass filter func.
def apply_lowpass_filter(signal, fs, target_rate):
    
    cutoff_hz = min(0.45 * fs, 20.0) if target_rate >= fs else target_rate / 2.0
    Wn = cutoff_hz / (fs / 2.0)
    Wn = min(max(Wn, 0.0001), 0.9999) 
    
    b, a = butter(N=4, Wn=Wn, btype='low')
    filtered = filtfilt(b, a, signal)
    
    print(f"Applied low-pass filter: {cutoff_hz:.2f} Hz cutoff")
    return filtered
```

```python
#Resampling Func.
def resample_signal(signal, duration_sec, target_rate):
 
    N_out = int(round(duration_sec * target_rate))
    resampled = resample(signal, N_out)
    
    print(f"Resampled: {len(signal)} -> {len(resampled)} samples at {target_rate} Hz")
    return resampled
```

```python
#Band-pass filter
def apply_bandpass_filter(signal, sampling_rate, lowcut=0.5, highcut=15.0):
   
    cleaned = nk.signal_filter(
        signal, lowcut=lowcut, highcut=highcut,
        method="butterworth", sampling_rate=sampling_rate, order=2
    )
    
    print(f"Applied band-pass filter: {lowcut}-{highcut} Hz")
    return cleaned
```

```python
import zipfile
import tempfile

def extract_h5_files_from_zip(zip_path):
    
    temp_dir = tempfile.mkdtemp()
    
    print(f"Extracting {zip_path}...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(temp_dir)
    
    # Find all H5 files
    h5_files = []
    for root, dirs, files in os.walk(temp_dir):
        for file in files:
            if file.endswith('.h5'):
                h5_files.append(os.path.join(root, file))
    
    print(f"Found {len(h5_files)} H5 files")
    return h5_files, temp_dir
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import find_peaks
import os

# Load your processed respiration data
processed_resp_data = pd.read_pickle(r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results\processed_respiration_data.pkl")

def calculate_respiratory_metrics(resp_signal, resp_time, sampling_rate=100):
    """
    Calculate detailed respiratory metrics from respiration signal
    """
    if len(resp_signal) < 10:  # Too short to analyze
        return {}
    
    # Find peaks (inspirations) - look for positive peaks
    peak_height = np.std(resp_signal) * 0.3  # Adaptive threshold
    min_distance = int(sampling_rate * 0.5)  # Minimum 0.5 seconds between breaths
    
    peaks, peak_properties = find_peaks(resp_signal, 
                                       height=peak_height, 
                                       distance=min_distance)
    
    # Find troughs (expirations) - look for negative peaks
    troughs, trough_properties = find_peaks(-resp_signal, 
                                          height=peak_height, 
                                          distance=min_distance)
    
    if len(peaks) < 2:  # Need at least 2 breaths
        return {
            'respiratory_rate_per_sec': 0,
            'respiratory_rate_per_min': 0,
            'interbreath_interval_mean': 0,
            'interbreath_interval_std': 0,
            'breath_count': 0,
            'breathing_amplitude_mean': 0,
            'breathing_amplitude_std': 0,
            'inspiration_duration_mean': 0,
            'expiration_duration_mean': 0,
            'respiratory_variability': 0
        }
    
    # Calculate interbreath intervals (peak to peak)
    peak_times = resp_time[peaks]
    interbreath_intervals = np.diff(peak_times)
    
    # Respiratory rate calculations
    total_duration = resp_time[-1] - resp_time[0]
    breath_count = len(peaks)
    respiratory_rate_per_sec = breath_count / total_duration if total_duration > 0 else 0
    respiratory_rate_per_min = respiratory_rate_per_sec * 60
    
    # Breathing amplitudes (peak to trough differences)
    amplitudes = []
    inspiration_durations = []
    expiration_durations = []
    
    for peak_idx in peaks:
        peak_value = resp_signal[peak_idx]
        
        # Find nearest trough before and after peak
        troughs_before = troughs[troughs < peak_idx]
        troughs_after = troughs[troughs > peak_idx]
        
        if len(troughs_before) > 0:
            trough_before_idx = troughs_before[-1]
            trough_before_value = resp_signal[trough_before_idx]
            amplitudes.append(peak_value - trough_before_value)
            
            # Inspiration duration (trough to peak)
            inspiration_duration = (peak_idx - trough_before_idx) / sampling_rate
            inspiration_durations.append(inspiration_duration)
        
        if len(troughs_after) > 0:
            trough_after_idx = troughs_after[0]
            
            # Expiration duration (peak to trough)
            expiration_duration = (trough_after_idx - peak_idx) / sampling_rate
            expiration_durations.append(expiration_duration)
    
    # Calculate metrics
    metrics = {
        'respiratory_rate_per_sec': respiratory_rate_per_sec,
        'respiratory_rate_per_min': respiratory_rate_per_min,
        'interbreath_interval_mean': np.mean(interbreath_intervals) if len(interbreath_intervals) > 0 else 0,
        'interbreath_interval_std': np.std(interbreath_intervals) if len(interbreath_intervals) > 0 else 0,
        'breath_count': breath_count,
        'breathing_amplitude_mean': np.mean(amplitudes) if len(amplitudes) > 0 else 0,
        'breathing_amplitude_std': np.std(amplitudes) if len(amplitudes) > 0 else 0,
        'inspiration_duration_mean': np.mean(inspiration_durations) if len(inspiration_durations) > 0 else 0,
        'expiration_duration_mean': np.mean(expiration_durations) if len(expiration_durations) > 0 else 0,
        'respiratory_variability': np.std(interbreath_intervals) / np.mean(interbreath_intervals) if len(interbreath_intervals) > 0 and np.mean(interbreath_intervals) > 0 else 0
    }
    
    return metrics

def extract_resp_during_bout(resp_data, bout_start_frame, bout_end_frame, video_fps, resp_fps=100):
    """
    Extract respiration data during a specific syllable bout
    """
    # Convert frame numbers to time
    start_time = bout_start_frame / video_fps
    end_time = bout_end_frame / video_fps
    
    # Convert time to respiration indices
    start_idx = int(start_time * resp_fps)
    end_idx = int(end_time * resp_fps)
    
    # Extract respiration during bout
    if start_idx < len(resp_data['respiration']) and end_idx < len(resp_data['respiration']):
        bout_resp = resp_data['respiration'][start_idx:end_idx]
        bout_time = resp_data['time'][start_idx:end_idx] - resp_data['time'][start_idx]  # Start from 0
        return bout_resp, bout_time
    else:
        return None, None

def analyze_resp_during_syllables(bout_details_file, processed_resp_data, min_bout_duration=0.5):
    """
    Analyze respiration patterns during syllable bouts with detailed metrics
    """
    # Load bout data
    bout_df = pd.read_excel(bout_details_file)
    
    # Filter out bouts shorter than 0.5 seconds
    print(f"Original bouts: {len(bout_df)}")
    filtered_bouts = bout_df[bout_df['bout_duration_seconds'] >= min_bout_duration]
    print(f"Filtered bouts (≥{min_bout_duration} seconds): {len(filtered_bouts)}")
    print(f"Removed {len(bout_df) - len(filtered_bouts)} short bouts")
    
    resp_bout_data = []
    
    for _, bout in filtered_bouts.iterrows():
        video_name = bout['video_name']
        
        # Find matching respiration file
        resp_key = f"{video_name}_merged.h5"
        
        if resp_key in processed_resp_data:
            resp_data = processed_resp_data[resp_key]
            video_fps = bout['video_fps']
            
            # Extract respiration during this bout
            bout_resp, bout_time = extract_resp_during_bout(
                resp_data, 
                bout['bout_start_frame'], 
                bout['bout_end_frame'], 
                video_fps
            )
            
            if bout_resp is not None:
                # Calculate detailed respiratory metrics
                resp_metrics = calculate_respiratory_metrics(bout_resp, bout_time)
                
                # Combine bout info with respiratory metrics
                bout_data = {
                    'syllable_id': bout['syllable_id'],
                    'video_name': bout['video_name'],
                    'track_number': bout['track_number'],
                    'bout_start_frame': bout['bout_start_frame'],
                    'bout_end_frame': bout['bout_end_frame'],
                    'bout_duration_sec': bout['bout_duration_seconds'],
                    'video_fps': bout['video_fps'],
                    'resp_mean': np.mean(bout_resp),
                    'resp_std': np.std(bout_resp),
                    'resp_range': np.max(bout_resp) - np.min(bout_resp),
                }
                
                # Add all respiratory metrics
                bout_data.update(resp_metrics)
                
                resp_bout_data.append(bout_data)
    
    return resp_bout_data

def create_detailed_resp_dataframe(resp_bout_data):
    """
    Create detailed DataFrame with all respiratory metrics
    """
    df = pd.DataFrame(resp_bout_data)
    
    # Reorder columns for better readability
    column_order = [
        'syllable_id', 'video_name', 'track_number', 
        'bout_start_frame', 'bout_end_frame', 'bout_duration_sec', 'video_fps',
        'respiratory_rate_per_sec', 'respiratory_rate_per_min', 
        'interbreath_interval_mean', 'interbreath_interval_std',
        'breath_count', 'breathing_amplitude_mean', 'breathing_amplitude_std',
        'inspiration_duration_mean', 'expiration_duration_mean', 
        'respiratory_variability', 'resp_mean', 'resp_std', 'resp_range'
    ]
    
    # Only include columns that exist
    available_columns = [col for col in column_order if col in df.columns]
    df = df[available_columns]
    
    return df

# Execute the analysis
bout_details_file = r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results\detailed_bout_frame_locations_correct_fps_FIXED.xlsx"

print("Analyzing respiration during syllable bouts (≥0.5 seconds)...")
resp_bout_data = analyze_resp_during_syllables(bout_details_file, processed_resp_data, min_bout_duration=0.5)

print(f"Found respiration data for {len(resp_bout_data)} syllable bouts")

# Create detailed DataFrame
detailed_resp_df = create_detailed_resp_dataframe(resp_bout_data)
print("\nFirst few rows of detailed respiration data:")
print(detailed_resp_df.head())

# Save results
output_dir = r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results"

# Save detailed data (like your screenshot)
detailed_file = os.path.join(output_dir, "detailed_respiration_syllable_analysis_filtered.xlsx")
detailed_resp_df.to_excel(detailed_file, index=False)

print(f"\nSaved detailed analysis to: {detailed_file}")
print("Analysis complete!")
```

```python
import numpy as np
import pandas as pd
from scipy.signal import find_peaks
import glob
import os

def process_respiratory_with_bandpass_filtering(resp_dir):
    processed_resp_data = pd.read_pickle(r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results\processed_respiration_data_0.5Hz.pkl")
    
    results = {}
    
    for file_id, data in processed_resp_data.items():
        print(f"Processing {file_id}")
        
        filtered_signal = data['respiration']
        fs = 100
        
        time_vector = np.arange(len(filtered_signal)) / fs
        
        normalized_signal = (filtered_signal - np.mean(filtered_signal)) / np.std(filtered_signal)
        peaks, _ = find_peaks(normalized_signal, distance=fs//4, prominence=0.5, height=0.1)
        
        print(f"Found {len(peaks)} breathing peaks")
        
        if len(peaks) < 2:
            print(f"Warning: Not enough peaks found")
            continue
            
        peak_times = peaks / fs
        ibi_seconds = np.diff(peak_times)
        
        breathing_rate_per_sec = 1 / ibi_seconds
        breathing_rate_bpm = breathing_rate_per_sec * 60
        
        breath_times = peak_times[1:]
        
        print(f"Mean breathing rate: {np.mean(breathing_rate_bpm):.1f} BPM")
        
        results[file_id] = {
            'filtered_signal': filtered_signal,
            'time_vector': time_vector,
            'breathing_peaks': peaks,
            'ibi_seconds': ibi_seconds,
            'breathing_rate_1_over_ibi': breathing_rate_per_sec,
            'breathing_rate_bpm': breathing_rate_bpm,
            'breath_times': breath_times,
            'sampling_rate': fs
        }
    
    return results

def get_fps_for_session(filename):
    fps_map = {
        "RI1_s1_1": 15,
        "RI1_s2_3": 15,
        "RI1_s3_6": 29,
        "RI2_s1_1": 15,
        "RI2_s2_4": 15
    }
    
    for pattern, fps in fps_map.items():
        if pattern in filename:
            return fps
    return 30

def overlay_resp_on_syllables_corrected(resp_data, excel_file_path):
    syllable_df = pd.read_excel(excel_file_path)
    print(f"Loaded syllable data: {syllable_df.shape}")
    
    video_fps = get_fps_for_session(os.path.basename(excel_file_path))
    syllable_times = syllable_df['frame_number'].values / video_fps
    
    print(f"Using {video_fps} FPS for time conversion")
    
    syllable_df['time_seconds'] = syllable_times
    syllable_df['fps_used'] = video_fps
    
    interp_resp = np.interp(syllable_times, resp_data['time_vector'], resp_data['filtered_signal'])
    
    if len(resp_data['breathing_rate_1_over_ibi']) > 0:
        interp_breathing_rate_1_over_ibi = np.interp(syllable_times, 
                                                    resp_data['breath_times'], 
                                                    resp_data['breathing_rate_1_over_ibi'])
        interp_breathing_rate_bpm = np.interp(syllable_times,
                                            resp_data['breath_times'],
                                            resp_data['breathing_rate_bpm'])
    else:
        interp_breathing_rate_1_over_ibi = np.nan * np.ones_like(syllable_times)
        interp_breathing_rate_bpm = np.nan * np.ones_like(syllable_times)
    
    syllable_df['resp_signal_bandpass_filtered'] = interp_resp
    syllable_df['breathing_rate_1_over_IBI'] = interp_breathing_rate_1_over_ibi
    syllable_df['breathing_rate_BPM'] = interp_breathing_rate_bpm
    syllable_df['IBI_seconds'] = 1 / interp_breathing_rate_1_over_ibi
    
    return syllable_df

def process_all_with_bandpass_data():
    resp_results = process_respiratory_with_bandpass_filtering("")
    
    excel_path = r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results"
    excel_files = glob.glob(os.path.join(excel_path, "*track0*.xlsx"))
    
    print(f"Found {len(excel_files)} Excel files")
    
    results = {}
    
    for file_id, resp_data in resp_results.items():
        print(f"Processing {file_id}")
        
        matching_excel = None
        for excel_file in excel_files:
            if 'RI1_s1_1' in file_id and 'RI1_s1_1' in excel_file:
                matching_excel = excel_file
                break
            elif 'RI1_s1_2' in file_id and 'RI1_s1_2' in excel_file:
                matching_excel = excel_file
                break
            elif 'RI1_s2_3' in file_id and 'RI1_s2_3' in excel_file:
                matching_excel = excel_file
                break
            elif 'RI1_s2_4' in file_id and 'RI1_s2_4' in excel_file:
                matching_excel = excel_file
                break
            elif 'RI1_s3_5' in file_id and 'RI1_s3_5' in excel_file:
                matching_excel = excel_file
                break
            elif 'RI1_s3_6' in file_id and 'RI1_s3_6' in excel_file:
                matching_excel = excel_file
                break
            elif 'RI1_s4_7' in file_id and 'RI1_s4_7' in excel_file:
                matching_excel = excel_file
                break
            elif 'RI1_s4_8' in file_id and 'RI1_s4_8' in excel_file:
                matching_excel = excel_file
                break
            elif 'RI1_s5_13' in file_id and 'RI1_s5_13' in excel_file:
                matching_excel = excel_file
                break
            elif 'RI2_s1_1' in file_id and 'RI2_s1_1' in excel_file:
                matching_excel = excel_file
                break
            elif 'RI2_s1_2' in file_id and 'RI2_s1_2' in excel_file:
                matching_excel = excel_file
                break
            elif 'RI2_s2_3' in file_id and 'RI2_s2_3' in excel_file:
                matching_excel = excel_file
                break
            elif 'RI2_s2_4' in file_id and 'RI2_s2_4' in excel_file:
                matching_excel = excel_file
                break
            elif 'RI2_s3_5' in file_id and 'RI2_s3_5' in excel_file:
                matching_excel = excel_file
                break
            elif 'RI2_s3_6' in file_id and 'RI2_s3_6' in excel_file:
                matching_excel = excel_file
                break
            elif 'RI2_s4_7' in file_id and 'RI2_s4_7' in excel_file:
                matching_excel = excel_file
                break
            elif 'RI2_s4_8' in file_id and 'RI2_s4_8' in excel_file:
                matching_excel = excel_file
                break
            elif 'RI2_s5_13' in file_id and 'RI2_s5_13' in excel_file:
                matching_excel = excel_file
                break
            
        if matching_excel:
            print(f"Matched with: {os.path.basename(matching_excel)}")
            overlaid_data = overlay_resp_on_syllables_corrected(resp_data, matching_excel)
            
            if overlaid_data is not None:
                excel_basename = os.path.basename(matching_excel).replace('.xlsx', '')
                output_file = f"CORRECTED_{excel_basename}_with_1_over_IBI.xlsx"
                
                overlaid_data.to_excel(output_file, index=False)
                print(f"Saved: {output_file}")
                
                valid_1_over_ibi = overlaid_data['breathing_rate_1_over_IBI'].dropna()
                if len(valid_1_over_ibi) > 0:
                    print(f"1/IBI range: {valid_1_over_ibi.min():.3f} - {valid_1_over_ibi.max():.3f} breaths/sec")
                
                results[file_id] = overlaid_data
        else:
            print(f"No matching Excel file found")
    
    return results

print("Processing with corrected 1/IBI calculation")
corrected_results = process_all_with_bandpass_data()

print(f"Processing complete - generated {len(corrected_results)} files")

if corrected_results:
    sample_key = list(corrected_results.keys())[0]
    sample_data = corrected_results[sample_key]
    
    print(f"Verification using {sample_key}")
    sample_1_over_ibi = sample_data['breathing_rate_1_over_IBI'].dropna().iloc[0]
    sample_ibi = sample_data['IBI_seconds'].dropna().iloc[0]
    print(f"1/IBI = {sample_1_over_ibi:.4f}, IBI = {sample_ibi:.4f}")
    print(f"Check: 1/{sample_ibi:.4f} = {1/sample_ibi:.4f}")
```

```python
import numpy as np
import pandas as pd
from scipy.signal import find_peaks
import glob
import os

def process_respiratory_bouts_with_bandpass_corrected():
    processed_resp_data = pd.read_pickle(r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results\processed_respiration_data_0.5Hz.pkl")
    
    excel_path = r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results"
    excel_files = glob.glob(os.path.join(excel_path, "*track0*.xlsx"))
    
    results = {}
    
    for file_id, resp_data in processed_resp_data.items():
        print(f"Processing {file_id}")
        
        filtered_signal = resp_data['respiration']
        fs = 100
        time_vector = np.arange(len(filtered_signal)) / fs
        
        normalized_signal = (filtered_signal - np.mean(filtered_signal)) / np.std(filtered_signal)
        peaks, _ = find_peaks(normalized_signal, distance=fs//4, prominence=0.5, height=0.1)
        
        print(f"Found {len(peaks)} breathing peaks")
        
        if len(peaks) < 2:
            continue
            
        peak_times = peaks / fs
        ibi_seconds = np.diff(peak_times)
        breathing_rate_1_over_ibi = 1 / ibi_seconds
        breathing_rate_bpm = breathing_rate_1_over_ibi * 60
        breath_times = peak_times[1:]
        
        print(f"IBI range: {ibi_seconds.min():.3f} - {ibi_seconds.max():.3f} seconds")
        print(f"Mean IBI: {ibi_seconds.mean():.3f} seconds")
        
        matching_excel = None
        for excel_file in excel_files:
            if any(session in file_id and session in excel_file for session in 
                   ['RI1_s1_1', 'RI1_s1_2', 'RI1_s2_3', 'RI1_s2_4', 'RI1_s3_5', 'RI1_s3_6', 
                    'RI1_s4_7', 'RI1_s4_8', 'RI1_s5_13', 'RI2_s1_1', 'RI2_s1_2', 'RI2_s2_3', 
                    'RI2_s2_4', 'RI2_s3_5', 'RI2_s3_6', 'RI2_s4_7', 'RI2_s4_8', 'RI2_s5_13']):
                matching_excel = excel_file
                break
        
        if matching_excel:
            print(f"Matched with: {os.path.basename(matching_excel)}")
            behavior_df = pd.read_excel(matching_excel)
            
            fps_map = {"RI1_s1_1": 15, "RI1_s2_3": 15, "RI1_s3_6": 29, "RI2_s1_1": 15, "RI2_s2_4": 15}
            video_fps = 30
            for pattern, fps in fps_map.items():
                if pattern in os.path.basename(matching_excel):
                    video_fps = fps
                    break
            
            print(f"Using {video_fps} FPS")
            
            behavior_df['time_seconds'] = behavior_df['frame_number'] / video_fps
            behavior_df['behavior_transition'] = (behavior_df['syllable'] != behavior_df['syllable'].shift()).astype(int)
            behavior_df['episode_group'] = behavior_df['behavior_transition'].cumsum()
            
            syllable_bouts = []
            
            for group_id, episode_group in behavior_df.groupby('episode_group'):
                syllable_id = episode_group['syllable'].iloc[0]
                
                bout_start_frame = episode_group['frame_number'].min()
                bout_end_frame = episode_group['frame_number'].max()
                bout_start_time_sec = episode_group['time_seconds'].min()
                bout_end_time_sec = episode_group['time_seconds'].max()
                bout_duration_ms = (bout_end_time_sec - bout_start_time_sec) * 1000
                bout_frame_count = len(episode_group)
                
                if bout_duration_ms >= 500:
                    
                    time_mask = (time_vector >= bout_start_time_sec) & (time_vector <= bout_end_time_sec)
                    bout_resp_signal = filtered_signal[time_mask]
                    
                    if len(bout_resp_signal) > 0:
                        mean_resp = np.mean(bout_resp_signal)
                        min_resp = np.min(bout_resp_signal)
                        max_resp = np.max(bout_resp_signal)
                        resp_range = max_resp - min_resp
                    else:
                        mean_resp = min_resp = max_resp = resp_range = np.nan
                    
                    breath_mask = (breath_times >= bout_start_time_sec) & (breath_times <= bout_end_time_sec)
                    bout_1_over_ibi = breathing_rate_1_over_ibi[breath_mask]
                    bout_bpm = breathing_rate_bpm[breath_mask]
                    bout_ibi = ibi_seconds[breath_mask]
                    
                    if len(bout_1_over_ibi) > 0:
                        mean_1_over_ibi = np.mean(bout_1_over_ibi)
                        mean_bpm = np.mean(bout_bpm)
                        mean_ibi = np.mean(bout_ibi)
                        breath_count = len(bout_1_over_ibi)
                    else:
                        mid_time = (bout_start_time_sec + bout_end_time_sec) / 2
                        if len(breath_times) > 0:
                            mean_1_over_ibi = np.interp(mid_time, breath_times, breathing_rate_1_over_ibi)
                            mean_bpm = mean_1_over_ibi * 60
                            mean_ibi = 1 / mean_1_over_ibi if mean_1_over_ibi > 0 else np.nan
                        else:
                            mean_1_over_ibi = mean_bpm = mean_ibi = np.nan
                        breath_count = 0
                    
                    syllable_bouts.append({
                        'syllable_id': syllable_id,
                        'bout_start_frame': bout_start_frame,
                        'bout_end_frame': bout_end_frame,
                        'bout_start_time_sec': bout_start_time_sec,
                        'bout_end_time_sec': bout_end_time_sec,
                        'bout_duration_ms': bout_duration_ms,
                        'bout_frame_count': bout_frame_count,
                        'fps_used': video_fps,
                        'mean_resp_signal_bandpass': mean_resp,
                        'min_resp_signal_bandpass': min_resp,
                        'max_resp_signal_bandpass': max_resp,
                        'resp_signal_range': resp_range,
                        'mean_1_over_IBI': mean_1_over_ibi,
                        'mean_breathing_rate_BPM': mean_bpm,
                        'mean_IBI_seconds': mean_ibi,
                        'breath_count': breath_count,
                        'video_name': episode_group['video_name'].iloc[0] if 'video_name' in episode_group.columns else 'unknown',
                        'track_name': episode_group['track_name'].iloc[0] if 'track_name' in episode_group.columns else 'track0'
                    })
            
            bouts_df = pd.DataFrame(syllable_bouts)
            bouts_df = bouts_df.sort_values('bout_start_time_sec').reset_index(drop=True)
            
            if len(bouts_df) > 0:
                excel_basename = os.path.basename(matching_excel).replace('.xlsx', '')
                output_file = f"FINAL_BOUTS_{excel_basename}.xlsx"
                bouts_df.to_excel(output_file, index=False)
                print(f"Saved: {output_file} with {len(bouts_df)} bouts")
                
                results[file_id] = bouts_df
        else:
            print(f"No matching file found")
    
    return results

def create_master_bout_files():
    bout_files = glob.glob("FINAL_BOUTS_*.xlsx")
    
    ri1_files = [f for f in bout_files if "-RI1_" in f]
    ri2_files = [f for f in bout_files if "-RI2_" in f]
    
    def create_master_file(files, output_name):
        with pd.ExcelWriter(output_name, engine='openpyxl') as writer:
            for file in files:
                df = pd.read_excel(file)
                
                filename = os.path.basename(file)
                if "-RI1_" in filename:
                    session_part = filename.split("-RI1_")[1]
                    sheet_prefix = "RI1_"
                elif "-RI2_" in filename:
                    session_part = filename.split("-RI2_")[1]
                    sheet_prefix = "RI2_"
                
                session_info = session_part.split("_")[0:3]
                sheet_name = f"{sheet_prefix}{'_'.join(session_info)}_track0"
                
                if len(sheet_name) > 31:
                    sheet_name = sheet_name[:31]
                
                df.to_excel(writer, sheet_name=sheet_name, index=False)
                print(f"Added {sheet_name}: {len(df)} bouts")
    
    if ri1_files:
        create_master_file(ri1_files, "RI1_MASTER_Bouts.xlsx")
    if ri2_files:
        create_master_file(ri2_files, "RI2_MASTER_Bouts.xlsx")

print("Processing respiratory bout data")
corrected_bout_results = process_respiratory_bouts_with_bandpass_corrected()
create_master_bout_files()
print(f"Done - processed {len(corrected_bout_results)} files")

if corrected_bout_results:
    sample_key = list(corrected_bout_results.keys())[0]
    sample_data = corrected_bout_results[sample_key]
    
    valid_ibi = sample_data['mean_IBI_seconds'].dropna()
    if len(valid_ibi) > 0:
        print(f"IBI range: {valid_ibi.min():.3f} - {valid_ibi.max():.3f} seconds")
        print(f"BPM range: {valid_ibi.min()*60:.1f} - {valid_ibi.max()*60:.1f}")
```

```python
import numpy as np
import pandas as pd
from scipy.signal import find_peaks
import glob
import os

def create_corrected_syllable_bout_files():
    processed_resp_data = pd.read_pickle(r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results\processed_respiration_data_0.5Hz.pkl")
    
    excel_path = r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results"
    excel_files = glob.glob(os.path.join(excel_path, "*track0*.xlsx"))
    
    results = {}
    
    for file_id, resp_data in processed_resp_data.items():
        print(f"Processing {file_id}")
        
        filtered_signal = resp_data['respiration']
        fs = 100
        time_vector = np.arange(len(filtered_signal)) / fs
        
        peaks, _ = find_peaks(filtered_signal, distance=fs * 0.0833)
        
        print(f"Found {len(peaks)} breathing peaks")
        
        if len(peaks) < 2:
            continue
            
        peak_times = peaks / fs
        ibi_seconds = np.diff(peak_times)
        breathing_rate_1_over_ibi = 1 / ibi_seconds
        breathing_rate_bpm = breathing_rate_1_over_ibi * 60
        breath_times = peak_times[1:]
        
        matching_excel = None
        for excel_file in excel_files:
            if any(session in file_id and session in excel_file for session in 
                   ['RI1_s1_1', 'RI1_s1_2', 'RI1_s2_3', 'RI1_s2_4', 'RI1_s3_5', 'RI1_s3_6', 
                    'RI1_s4_7', 'RI1_s4_8', 'RI1_s5_13', 'RI2_s1_1', 'RI2_s1_2', 'RI2_s2_3', 
                    'RI2_s2_4', 'RI2_s3_5', 'RI2_s3_6', 'RI2_s4_7', 'RI2_s4_8', 'RI2_s5_13']):
                matching_excel = excel_file
                break
        
        if matching_excel:
            print(f"Matched with: {os.path.basename(matching_excel)}")
            behavior_df = pd.read_excel(matching_excel)
            
            fps_map = {"RI1_s1_1": 15, "RI1_s2_3": 15, "RI1_s3_6": 29, "RI2_s1_1": 15, "RI2_s2_4": 15}
            video_fps = 30
            for pattern, fps in fps_map.items():
                if pattern in os.path.basename(matching_excel):
                    video_fps = fps
                    break
            
            behavior_df['time_seconds'] = behavior_df['frame_number'] / video_fps
            behavior_df['behavior_transition'] = (behavior_df['syllable'] != behavior_df['syllable'].shift()).astype(int)
            behavior_df['episode_group'] = behavior_df['behavior_transition'].cumsum()
            
            syllable_bouts = []
            
            for group_id, episode_group in behavior_df.groupby('episode_group'):
                syllable_id = episode_group['syllable'].iloc[0]
                
                bout_start_frame = episode_group['frame_number'].min()
                bout_end_frame = episode_group['frame_number'].max()
                bout_start_time_sec = episode_group['time_seconds'].min()
                bout_end_time_sec = episode_group['time_seconds'].max()
                bout_duration_ms = (bout_end_time_sec - bout_start_time_sec) * 1000
                bout_frame_count = len(episode_group)
                
                if bout_duration_ms >= 500:
                    
                    time_mask = (time_vector >= bout_start_time_sec) & (time_vector <= bout_end_time_sec)
                    bout_resp_signal = filtered_signal[time_mask]
                    
                    if len(bout_resp_signal) > 0:
                        mean_resp = np.mean(bout_resp_signal)
                        min_resp = np.min(bout_resp_signal)
                        max_resp = np.max(bout_resp_signal)
                        resp_range = max_resp - min_resp
                    else:
                        mean_resp = min_resp = max_resp = resp_range = np.nan
                    
                    breath_mask = (breath_times >= bout_start_time_sec) & (breath_times <= bout_end_time_sec)
                    bout_1_over_ibi = breathing_rate_1_over_ibi[breath_mask]
                    bout_bpm = breathing_rate_bpm[breath_mask]
                    bout_ibi = ibi_seconds[breath_mask]
                    
                    if len(bout_1_over_ibi) > 0:
                        mean_1_over_ibi = np.mean(bout_1_over_ibi)
                        mean_bpm = np.mean(bout_bpm)
                        mean_ibi = np.mean(bout_ibi)
                        breath_count = len(bout_1_over_ibi)
                    else:
                        mid_time = (bout_start_time_sec + bout_end_time_sec) / 2
                        if len(breath_times) > 0:
                            mean_1_over_ibi = np.interp(mid_time, breath_times, breathing_rate_1_over_ibi)
                            mean_bpm = mean_1_over_ibi * 60
                            mean_ibi = 1 / mean_1_over_ibi if mean_1_over_ibi > 0 else np.nan
                        else:
                            mean_1_over_ibi = mean_bpm = mean_ibi = np.nan
                        breath_count = 0
                    
                    syllable_bouts.append({
                        'syllable_id': syllable_id,
                        'bout_start_frame': bout_start_frame,
                        'bout_end_frame': bout_end_frame,
                        'bout_start_time_sec': bout_start_time_sec,
                        'bout_end_time_sec': bout_end_time_sec,
                        'bout_duration_ms': bout_duration_ms,
                        'bout_frame_count': bout_frame_count,
                        'fps_used': video_fps,
                        'mean_resp_signal_bandpass': mean_resp,
                        'min_resp_signal_bandpass': min_resp,
                        'max_resp_signal_bandpass': max_resp,
                        'resp_signal_range': resp_range,
                        'mean_1_over_IBI': mean_1_over_ibi,
                        'mean_breathing_rate_BPM': mean_bpm,
                        'mean_IBI_seconds': mean_ibi,
                        'breath_count': breath_count,
                        'video_name': episode_group['video_name'].iloc[0] if 'video_name' in episode_group.columns else 'unknown',
                        'track_name': episode_group['track_name'].iloc[0] if 'track_name' in episode_group.columns else 'track0'
                    })
            
            bouts_df = pd.DataFrame(syllable_bouts)
            bouts_df = bouts_df.sort_values('bout_start_time_sec').reset_index(drop=True)
            
            if len(bouts_df) > 0:
                excel_basename = os.path.basename(matching_excel).replace('.xlsx', '')
                output_file = f"CORRECTED_PEAKS_{excel_basename}.xlsx"
                bouts_df.to_excel(output_file, index=False)
                print(f"Saved: {output_file} with {len(bouts_df)} bouts")
                
                results[file_id] = bouts_df
    
    return results

def create_corrected_master_files():
    bout_files = glob.glob("CORRECTED_PEAKS_*.xlsx")
    
    ri1_files = [f for f in bout_files if "-RI1_" in f]
    ri2_files = [f for f in bout_files if "-RI2_" in f]
    
    print(f"Found {len(ri1_files)} RI1 files and {len(ri2_files)} RI2 files")
    
    def create_master_file(files, output_name):
        with pd.ExcelWriter(output_name, engine='openpyxl') as writer:
            for file in files:
                df = pd.read_excel(file)
                
                filename = os.path.basename(file)
                if "-RI1_" in filename:
                    session_part = filename.split("-RI1_")[1]
                    sheet_prefix = "RI1_"
                elif "-RI2_" in filename:
                    session_part = filename.split("-RI2_")[1]
                    sheet_prefix = "RI2_"
                
                session_info = session_part.split("_")[0:3]
                sheet_name = f"{sheet_prefix}{'_'.join(session_info)}_track0"
                
                if len(sheet_name) > 31:
                    sheet_name = sheet_name[:31]
                
                df.to_excel(writer, sheet_name=sheet_name, index=False)
                print(f"Added {sheet_name}: {len(df)} bouts")
    
    if ri1_files:
        create_master_file(ri1_files, "RI1_MASTER_Corrected_Peaks.xlsx")
    if ri2_files:
        create_master_file(ri2_files, "RI2_MASTER_Corrected_Peaks.xlsx")

print("Creating corrected Excel files with improved peak detection")
corrected_results = create_corrected_syllable_bout_files()
create_corrected_master_files()

print(f"Completed - processed {len(corrected_results)} files")
print("Created master files:")
print("- RI1_MASTER_Corrected_Peaks.xlsx")
print("- RI2_MASTER_Corrected_Peaks.xlsx")

if corrected_results:
    sample_key = list(corrected_results.keys())[0]
    sample_data = corrected_results[sample_key]
    
    valid_ibi = sample_data['mean_IBI_seconds'].dropna()
    valid_1_over_ibi = sample_data['mean_1_over_IBI'].dropna()
    
    if len(valid_ibi) > 0:
        print(f"\nData verification:")
        print(f"IBI range: {valid_ibi.min():.3f} - {valid_ibi.max():.3f} seconds")
        print(f"1/IBI range: {valid_1_over_ibi.min():.3f} - {valid_1_over_ibi.max():.3f} breaths/sec")
        print(f"BPM range: {valid_1_over_ibi.min()*60:.1f} - {valid_1_over_ibi.max()*60:.1f}")
        
        sample_ibi = valid_ibi.iloc[0]
        sample_1_over_ibi = valid_1_over_ibi.iloc[0]
        print(f"Math check: IBI={sample_ibi:.3f}s, 1/IBI={sample_1_over_ibi:.3f}, 1/{sample_ibi:.3f}={1/sample_ibi:.3f}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import find_peaks

def verify_multiple_syllables():
    """Verify multiple syllables across different sessions"""
    
    # Test different files and rows
    test_cases = [
        ("CORRECTED_PEAK_COUNT_top-RI1_s1_1_p5_2_nRB6_20250622_143958_track0_syllables.xlsx", "RI1_s1_1", [2, 5, 10, 15]),
        ("CORRECTED_PEAK_COUNT_top-RI1_s2_3_p5_3_nRB3_20250622_104059_track0_syllables.xlsx", "RI1_s2_3", [1, 8, 12, 20]),
        ("CORRECTED_PEAK_COUNT_top-RI1_s3_6_p5_3_nRB3_20250621_125312_track0_syllables.xlsx", "RI1_s3_6", [3, 7, 14, 18])
    ]
    
    processed_resp_data = pd.read_pickle(r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results\processed_respiration_data_0.5Hz.pkl")
    
    def graph_peak_detection(signal, fs=100):
        signal_std = np.std(signal)
        peaks, _ = find_peaks(signal,
                             distance=fs * 0.08,
                             prominence=signal_std * 0.2,
                             height=np.mean(signal))
        return peaks
    
    for excel_file, session_pattern, test_rows in test_cases:
        if not os.path.exists(excel_file):
            print(f"File not found: {excel_file}")
            continue
            
        print(f"\n=== TESTING {session_pattern} ===")
        
        # Load data
        df = pd.read_excel(excel_file)
        session_key = [key for key in processed_resp_data.keys() if session_pattern in key][0]
        sig = processed_resp_data[session_key]['respiration']
        fs = 100
        
        # Detect peaks
        peaks = graph_peak_detection(sig, fs)
        peak_times = peaks / fs
        
        fig, axes = plt.subplots(2, 2, figsize=(16, 10))
        axes = axes.flatten()
        
        matches = 0
        total = 0
        
        for i, row_idx in enumerate(test_rows[:4]):  # Max 4 plots
            if row_idx >= len(df):
                continue
                
            bout = df.iloc[row_idx]
            exact_start = bout['bout_start_time_sec']
            exact_end = bout['bout_end_time_sec']
            
            # Count peaks in syllable
            syllable_peak_mask = (peak_times >= exact_start) & (peak_times <= exact_end)
            syllable_peaks = peak_times[syllable_peak_mask]
            graph_count = len(syllable_peaks)
            excel_count = bout['breath_count']
            
            match = (graph_count == excel_count)
            matches += match
            total += 1
            
            # Plot
            context = 1.0
            plot_start = max(0, exact_start - context)
            plot_end = min(len(sig)/fs, exact_end + context)
            
            i0 = int(plot_start * fs)
            i1 = int(plot_end * fs)
            t_plot = np.arange(i0, i1) / fs
            
            window_peaks = peak_times[(peak_times >= plot_start) & (peak_times <= plot_end)]
            
            ax = axes[i]
            ax.plot(t_plot, sig[i0:i1], 'b-', lw=1.5, label="Signal")
            ax.plot(window_peaks, sig[(window_peaks * fs).astype(int)], 'ro', ms=3, alpha=0.6, label="All peaks")
            ax.plot(syllable_peaks, sig[(syllable_peaks * fs).astype(int)], 'go', ms=6, 
                   label=f"Syllable peaks ({len(syllable_peaks)})")
            
            ax.axvline(exact_start, color='red', linestyle='-', linewidth=2, alpha=0.8)
            ax.axvline(exact_end, color='red', linestyle='-', linewidth=2, alpha=0.8)
            ax.axvspan(exact_start, exact_end, alpha=0.3, color='yellow')
            
            ax.set_xlabel("Time (s)")
            ax.set_ylabel("Signal")
            ax.set_title(f"Row {row_idx}: Syl {bout['syllable_id']} - Graph:{graph_count} Excel:{excel_count} {'✓' if match else '✗'}")
            ax.legend(fontsize=8)
            ax.grid(True, alpha=0.3)
            
            print(f"  Row {row_idx}: Syllable {bout['syllable_id']} - Graph:{graph_count} Excel:{excel_count} {'✓' if match else '✗'}")
        
        plt.suptitle(f"{session_pattern} Verification - {matches}/{total} matches", fontsize=14)
        plt.tight_layout()
        plt.show()
        
        print(f"  Summary: {matches}/{total} syllables match")

# Run verification on multiple files
verify_multiple_syllables()
```

```python
def test_specific_time_windows():
    """Test specific time windows across different sessions"""
    
    time_tests = [
        ("RI1_s1_1", 50.0, 55.0, "Early recording"),
        ("RI1_s1_1", 150.0, 155.0, "Mid recording"),  
        ("RI1_s2_3", 100.0, 105.0, "Different session"),
        ("RI1_s3_6", 200.0, 205.0, "29 FPS session"),
        ("RI2_s1_1", 80.0, 85.0, "RI2 example")
    ]
    
    processed_resp_data = pd.read_pickle(r"C:\Users\Padilla-Coreano\Documents\tanish\Keypoint RI1+RI2\Excel_Results\processed_respiration_data_0.5Hz.pkl")
    
    def graph_peak_detection(signal, fs=100):
        signal_std = np.std(signal)
        peaks, _ = find_peaks(signal,
                             distance=fs * 0.08,
                             prominence=signal_std * 0.2,
                             height=np.mean(signal))
        return peaks
    
    fig, axes = plt.subplots(3, 2, figsize=(15, 12))
    axes = axes.flatten()
    
    for i, (session, start_time, end_time, description) in enumerate(time_tests):
        if i >= 6:  # Max 6 plots
            break
            
        # Find session key
        session_key = None
        for key in processed_resp_data.keys():
            if session in key:
                session_key = key
                break
        
        if session_key is None:
            continue
            
        sig = processed_resp_data[session_key]['respiration']
        fs = 100
        
        # Detect peaks
        peaks = graph_peak_detection(sig, fs)
        peak_times = peaks / fs
        
        # Plot time window
        i0 = int(start_time * fs)
        i1 = int(end_time * fs)
        t_plot = np.arange(i0, i1) / fs
        
        window_peaks = peak_times[(peak_times >= start_time) & (peak_times <= end_time)]
        
        ax = axes[i]
        ax.plot(t_plot, sig[i0:i1], 'b-', lw=1.5, label="Signal")
        ax.plot(window_peaks, sig[(window_peaks * fs).astype(int)], 'ro', ms=5, 
               label=f"Peaks ({len(window_peaks)})")
        
        ax.set_xlabel("Time (s)")
        ax.set_ylabel("Signal")
        ax.set_title(f"{session} ({start_time}-{end_time}s) - {description}")
        ax.legend()
        ax.grid(True, alpha=0.3)
        
        # Calculate BPM for this window
        if len(window_peaks) > 1:
            window_ibis = np.diff(window_peaks)
            window_bpm = (1/np.mean(window_ibis)) * 60
            ax.text(0.02, 0.95, f'{window_bpm:.0f} BPM', transform=ax.transAxes,
                   bbox=dict(boxstyle="round,pad=0.3", facecolor="white", alpha=0.8))
    
    plt.suptitle("Peak Detection Across Different Sessions and Time Windows", fontsize=14)
    plt.tight_layout()
    plt.show()

# Run time window tests
test_specific_time_windows()
```
